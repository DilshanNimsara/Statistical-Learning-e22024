# Q Bayesian Estimations for Structural Health Monitoring via Bounded Grid Updates

In aerospace and civil engineering, Structural Health Monitoring (SHM) is critical for detecting damage before a catastrophic failure occurs. Consider an aircraft wing or a bridge girder equipped with specialized vibration sensors. Over time, environmental fatigue or dynamic impacts can cause micro-fractures, resulting in a reduction of the component's mechanical stiffness.

Let $\Theta = \theta$ represent the structural **remaining stiffness efficiency factor**, where $\theta$ is physically bounded to the interval:

$$\theta \in (0, 1]$$

* $\theta = 1.0$ indicates a perfectly pristine, undamaged structural component.
* $\theta \to 0$ signifies critical degradation or severe structural cracking.

Let $K_{\text{nominal}}$ be the known, baseline stiffness of the structural component when it is entirely healthy. At each sequential inspection time step $k$ (where $k = 1, 2, \dots, n$), a sensor collects a noisy experimental stiffness measurement $y_k$.

Engineers model the degradation physics via a non-linear relationship with multiplicative log-normal measurement noise to prevent non-physical negative values:

$$y_k = \theta \cdot K_{\text{nominal}} \cdot e^{\epsilon_k}, \qquad \epsilon_k \sim \mathscr{N}(0, \sigma^2)$$

where $\sigma$ is the standard deviation of the sensor noise in log-space.

Let $\mathbf{y}^{(k)} = (y_1, y_2, \dots, y_k)$ represent the **running history vector of observed sensor readings** up to the current inspection milestone. Before deploying the sensors, engineers utilize an initial prior distribution $f_{\Theta}^{(0)}(\theta)$ over the domain $(0, 1]$ based on historical manufacturing specifications. As the sensor stream arrives, the posterior distribution calculated at step $k-1$ serves directly as the prior distribution for step $k$.

## Sample Answer

1. Prior Belief BoundariesExpected Prior ValueFor a Beta-distributed random variable $\Theta \sim \text{Beta}(\alpha, \beta)$ with parameters $\alpha = 8$ and $\beta = 1.5

$:$$\mathbb{E}\left[\Theta^{(0)}\right] = \frac{\alpha}{\alpha + \beta} = \frac{8}{8 + 1.5} = \frac{8}{9.5} \approx 0.8421$$

Engineering Justification

Optimistic Bias Towards Structural Health: In Structural Health Monitoring (SHM), newly manufactured or recently inspected structural components are assumed to be non-degraded until sensor evidence indicates otherwise.

Physical Domain Bounds:

The Beta distribution natively bounds $\theta \in (0, 1]$, strictly enforcing physically realistic parameters ($\theta = 1.0$ represents pristine nominal stiffness $K_{\text{nominal}}$, while $\theta \to 0$ represents severe damage/cracking) without allowing invalid negative stiffness values.

Skewness Profile:

With $\alpha = 8 > \beta = 1.5$, the distribution places strong probability mass near $1.0$, allowing early sensor noise to be smoothed out while still remaining open to updating if severe structural degradation ($\theta \ll 1.0$) is observed.

2. Structural Likelihood Formulation

Given the log-normal forward response model for a continuous measurement $y_k$:

$$y_k = \theta \cdot K_{\text{nominal}} \cdot e^{\epsilon_k}, \quad \epsilon_k \sim \mathcal{N}(0, \sigma^2)$$

Taking the natural logarithm of both sides yields:

$$\ln(y_k) = \ln(\theta \cdot K_{\text{nominal}}) + \epsilon_k \implies \ln(y_k) \sim \mathcal{N}\left(\ln(\theta \cdot K_{\text{nominal}}), \sigma^2\right)$$

Single Measurement Likelihood $L(y_k \mid \theta)$Using the log-normal probability density function (or change of variables $d\ln(y_k) = \frac{1}{y_k} dy_k$):

$$L(y_k \mid \theta) = \frac{1}{y_k \sigma \sqrt{2\pi}} \exp\left( -\frac{\left( \ln(y_k) - \ln(\theta \cdot K_{\text{nominal}}) \right)^2}{2\sigma^2} \right)$$

Joint Likelihood Function $L(\mathbf{y}^{(k)} \mid \theta)$

Assuming conditional independence of sensor noise across steps $1, \dots, k$:

$$L(\mathbf{y}^{(k)} \mid \theta) = \prod_{i=1}^k L(y_i \mid \theta) = \left( \frac{1}{\sigma \sqrt{2\pi}} \right)^k \left( \prod_{i=1}^k \frac{1}{y_i} \right) \exp\left( -\frac{1}{2\sigma^2} \sum_{i=1}^k \left( \ln(y_i) - \ln(\theta \cdot K_{\text{nominal}}) \right)^2 \right)$$

3. Non-Conjugate Grid Update FormulationNon-Conjugacy Explanation

A closed-form analytical solution for the posterior density $f_{\Theta \mid \mathbf{Y}^{(k)}}(\theta \mid \mathbf{y}^{(k)})$ does not exist because:

The initial prior is a Beta distribution over $\theta \in (0, 1]$.The structural observation likelihood follows a Log-Normal distribution with parameter $\theta$ embedded non-linearly inside the scale parameter $\ln(\theta \cdot K_{\text{nominal}})$.

Since the Beta family is not conjugate to the Log-Normal likelihood, multiplying the density functions yields an intractable posterior whose normalizing denominator $Z_k$ cannot be evaluated in closed form.

Recursive Posterior RelationshipAt step $k$, using the step $k-1$ posterior $f_{\Theta \mid \mathbf{Y}^{(k-1)}}(\theta \mid \mathbf{y}^{(k-1)})$ as the prior for step $k$:

$$f_{\Theta \mid \mathbf{Y}^{(k)}}(\theta \mid \mathbf{y}^{(k)}) \propto L(y_k \mid \theta) \, f_{\Theta \mid \mathbf{Y}^{(k-1)}}(\theta \mid \mathbf{y}^{(k-1)})$$

Explicitly:

$$f_{\Theta \mid \mathbf{Y}^{(k)}}(\theta \mid \mathbf{y}^{(k)}) = \frac{1}{Z_k} \cdot \left[ \frac{1}{y_k \sigma \sqrt{2\pi}} \exp\left( -\frac{\left(\ln y_k - \ln(\theta K_{\text{nominal}})\right)^2}{2\sigma^2} \right) \right] \cdot f_{\Theta \mid \mathbf{Y}^{(k-1)}}(\theta \mid \mathbf{y}^{(k-1)})$$where the normalizing constant is:$$Z_k = \int_{0.01}^{1.0} L(y_k \mid \theta) \, f_{\Theta \mid \mathbf{Y}^{(k-1)}}(\theta \mid \mathbf{y}^{(k-1)}) \, d\theta$$

4. Running Point Estimators

Over the bounded domain $\theta \in [0.01, 1.0]$:

a. Running Posterior Mean ($\widehat{\theta}_{\text{Bayes}}^{(k)}$)$$\widehat{\theta}_{\text{Bayes}}^{(k)} = \int_{0.01}^{1.0} \theta \, f_{\Theta \mid \mathbf{Y}^{(k)}}(\theta \mid \mathbf{y}^{(k)}) \, d\theta$$

b. Running Maximum A Posteriori ($\widehat{\theta}_{\text{MAP}}^{(k)}$)$$\widehat{\theta}_{\text{MAP}}^{(k)} = \arg\max_{\theta \in [0.01, 1.0]} f_{\Theta \mid \mathbf{Y}^{(k)}}(\theta \mid \mathbf{y}^{(k)})$$

5. Algorithmic Grid Approximation and NormalizationGrid Setup:

Discretize the physical interval $[0.01, 1.0]$ into $M$ linearly spaced points:

$$\theta_m = 0.01 + (m-1)\Delta\theta, \quad \Delta\theta = \frac{1.0 - 0.01}{M-1}, \quad m = 1, \dots, M$$

Boundary enforcement: Values below $0.01$ or above $1.0$ are truncated/zeroed.Initialization ($k=0$):Evaluate the initial Beta prior array:

$$\tilde{P}_0(\theta_m) = \text{Beta.pdf}(\theta_m; a=8, b=1.5)$$

Normalize using the composite trapezoidal rule:

$$Z_0 = \text{trapezoid}(\tilde{\mathbf{P}}_0, \boldsymbol{\theta}) \implies \mathbf{P}_0 = \frac{\tilde{\mathbf{P}}_0}{Z_0}$$

Sequential Processing Step $k$:

Compute likelihood across grid:

$L(y_k \mid \theta_m) = \text{lognorm.pdf}\left(y_k, \sigma=\sigma, \text{scale}=\theta_m \cdot K_{\text{nominal}}\right)$

Unnormalized update: $\tilde{P}_k(\theta_m) = P_{k-1}(\theta_m) \times L(y_k \mid \theta_m)$

Compute normalizing constant via trapezoidal rule:$$Z_k = \sum_{m=1}^{M-1} \frac{\tilde{P}_k(\theta_m) + \tilde{P}_k(\theta_{m+1})}{2} \Delta\theta$$Posterior normalization: $P_k(\theta_m) = \frac{\tilde{P}_k(\theta_m)}{Z_k}$

Point Estimation:

$\widehat{\theta}_{\text{Bayes}}^{(k)} \approx \text{trapezoid}(\boldsymbol{\theta} \odot \mathbf{P}_k, \boldsymbol{\theta})$$\widehat{\theta}_{\text{MAP}}^{(k)} = \theta_{m^*}, \quad m^* = \arg\max_{m} P_k(\theta_m)$

6. Simulation & Performance Analysis Python Code

```python
import numpy as np
import scipy.stats as stats
import plotly.graph_objects as go

# Set random seed for reproducibility
np.random.seed(42)

# Parameters
theta_true = 0.68
K_nominal = 50.0  # kN/mm
sigma = 0.15
n_steps = 15
M = 500  # Number of grid points

# 1. Grid Discretization over bounded domain [0.01, 1.0]
theta_grid = np.linspace(0.01, 1.0, M)

# 2. Prior Initialization: Beta(8, 1.5)
P_k = stats.beta.pdf(theta_grid, a=8, b=1.5)
P_k /= np.trapezoid(P_k, theta_grid)  # Initial normalization

# Storage arrays
milestones = [0, 1, 2, 5, 10, 15]
density_profiles = {0: P_k.copy()}

bayes_estimates = [np.trapezoid(theta_grid * P_k, theta_grid)]
map_estimates = [theta_grid[np.argmax(P_k)]]

# Sequential Simulation Loop
for k in range(1, n_steps + 1):
    # Simulate Sensor Observation y_k ~ LogNormal(meanlog=ln(theta_true * K_nominal), sdlog=sigma)
    noise = np.random.normal(0, sigma)
    y_k = (theta_true * K_nominal) * np.exp(noise)

    # Evaluate Likelihood across grid
    # Scale parameter in scipy lognorm is exp(mu) = theta_grid * K_nominal
    likelihood = stats.lognorm.pdf(y_k, s=sigma, scale=theta_grid * K_nominal)

    # Bayesian Update
    P_unnorm = P_k * likelihood
    Z_k = np.trapezoid(P_unnorm, theta_grid)
    P_k = P_unnorm / Z_k

    # Point Estimates
    theta_bayes = np.trapezoid(theta_grid * P_k, theta_grid)
    theta_map = theta_grid[np.argmax(P_k)]

    bayes_estimates.append(theta_bayes)
    map_estimates.append(theta_map)

    if k in milestones:
        density_profiles[k] = P_k.copy()

# Visualizations

# Figure 1: Posterior Density Profiles Progression
fig1 = go.Figure()
for k_step, density in density_profiles.items():
    dash_style = 'dash' if k_step == 0 else 'solid'
    fig1.add_trace(go.Scatter(
        x=theta_grid, y=density, mode='lines',
        name=f"Step {k_step}" if k_step > 0 else "Prior (Beta)",
        line=dict(dash=dash_style, width=2)
    ))

fig1.add_vline(x=theta_true, line_dash="dot", line_color="red", line_width=2,
               annotation_text=f"True State ({theta_true})")
fig1.update_layout(
    title="Posterior Density Progression Across Inspection Milestones",
    xaxis_title="Remaining Stiffness Factor (θ)",
    yaxis_title="Probability Density",
    template="plotly_white"
)
fig1.show()

# Figure 2: Convergence Timeline
fig2 = go.Figure()
steps = list(range(n_steps + 1))
fig2.add_trace(go.Scatter(x=steps, y=bayes_estimates, mode='lines+markers', name="Posterior Mean (Bayes)"))
fig2.add_trace(go.Scatter(x=steps, y=map_estimates, mode='lines+markers', name="MAP Estimator"))
fig2.add_hline(y=theta_true, line_dash="dash", line_color="red", annotation_text="True Damage State (0.68)")

fig2.update_layout(
    title="Estimator Convergence Timeline vs. True Degradation State",
    xaxis_title="Inspection Step (k)",
    yaxis_title="Stiffness Factor Estimate (θ)",
    template="plotly_white"
)
fig2.show()
```
