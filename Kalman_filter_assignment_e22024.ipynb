# Q. Analytical Derivation

Consider the filter model
\begin{align*}
x^{-}_k &= A_{k-1}\,x^{+}_{k-1} + G_{k-1}\,w_{k-1}, \\
y^{-}_k &= H_k\,x^{-}_k + z_k,
\end{align*}
where
$$x^{+}_{k-1} \sim \mathscr{N}(m_{k-1}, P_{k-1}),$$

Answer the followings:

1. Show that
$$x_k^- \sim \mathscr{N}(m_k^-, P_k^-),$$
where
$$m_k^- = A_{k-1}m_{k-1},$$
and
$$
P_k^- = A_{k-1}P_{k-1}A_{k-1}^T + G_{k-1}\Sigma_p G_{k-1}^T.
$$

2. Show that
$$y_k^- \sim \mathscr{N}(H_k m_k^-, H_k P_k^- H_k^T + \Sigma_m).$$

3. Show that
$$\begin{bmatrix} x_k^- \\ y^{-}_k \end{bmatrix} \sim \mathscr{N}\left(
\begin{bmatrix} m_k^- \\ H_k m_k^- \end{bmatrix},
\begin{bmatrix} P_k^- & P_k^- H_k^T \\ H_k P_k^- & H_k P_k^- H_k^T + \Sigma_m \end{bmatrix}
\right).$$

4. Show that
$$x^{+}_k\triangleq (x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k}) \sim \mathscr{N}(m_k, P_k),$$
where the updated mean $m_k$ and updated covariance $P_k$ are given by:
\begin{align*}
K_k &\triangleq P_k^- H_k^T (H_k P_k^- H_k^T + \Sigma_m)^{-1}, \\
m_k &= m_k^- + K_k (y^{\mathrm{obs}}_{k} - H_k m_k^-), \\
P_k &= (I - K_k H_k) P_k^-.
\end{align*}

5. Find
$\mathbb{E}[x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k}]$  and
$\text{Var}(x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k})$.

## Answers
## 1. Distribution of the Predicted State $x^-_k$

We are given the state prediction equation:


$$x^-_k = A_{k-1}x^+_{k-1} + G_{k-1}w_{k-1}$$

Since $x^+_{k-1} \sim \mathcal{N}(m_{k-1}, P_{k-1})$ and $w_{k-1} \sim \mathcal{N}(0, \Sigma_p)$ are independent Gaussian random variables, any linear combination of them is also Gaussian. Thus, $x^-_k \sim \mathcal{N}(m^-_k, P^-_k)$.

### Expected Value (Mean)

Using the linearity of expectation:


$$m^-_k = \mathbb{E}[x^-_k] = \mathbb{E}[A_{k-1}x^+_{k-1} + G_{k-1}w_{k-1}]$$

$$m^-_k = A_{k-1}\mathbb{E}[x^+_{k-1}] + G_{k-1}\mathbb{E}[w_{k-1}]$$

Since $\mathbb{E}[x^+_{k-1}] = m_{k-1}$ and $\mathbb{E}[w_{k-1}] = 0$:


$$m^-_k = A_{k-1}m_{k-1}$$

### Covariance

Using the property $\text{Var}(BX + CY) = B\text{Var}(X)B^T + C\text{Var}(Y)C^T$ for independent variables $X$ and $Y$:


$$P^-_k = \text{Var}(x^-_k) = \text{Var}(A_{k-1}x^+_{k-1} + G_{k-1}w_{k-1})$$

$$P^-_k = A_{k-1}\text{Var}(x^+_{k-1})A_{k-1}^T + G_{k-1}\text{Var}(w_{k-1})G_{k-1}^T$$

Substituting $\text{Var}(x^+_{k-1}) = P_{k-1}$ and $\text{Var}(w_{k-1}) = \Sigma_p$:


$$P^-_k = A_{k-1}P_{k-1}A_{k-1}^T + G_{k-1}\Sigma_p G_{k-1}^T$$

---

## 2. Distribution of the Predicted Measurement $y^-_k$

The measurement equation is:


$$y^-_k = H_k x^-_k + z_k$$

Since $x^-_k \sim \mathcal{N}(m^-_k, P^-_k)$ and $z_k \sim \mathcal{N}(0, \Sigma_m)$ are independent, $y^-_k$ is also normally distributed: $y^-_k \sim \mathcal{N}(\mathbb{E}[y^-_k], \text{Var}(y^-_k))$.

### Expected Value (Mean)

$$\mathbb{E}[y^-_k] = \mathbb{E}[H_k x^-_k + z_k] = H_k\mathbb{E}[x^-_k] + \mathbb{E}[z_k]$$

$$\mathbb{E}[y^-_k] = H_k m^-_k + 0 = H_k m^-_k$$

### Covariance

$$P_{yy} = \text{Var}(y^-_k) = \text{Var}(H_k x^-_k + z_k)$$

$$P_{yy} = H_k \text{Var}(x^-_k) H_k^T + \text{Var}(z_k) = H_k P^-_k H_k^T + \Sigma_m$$

Thus:


$$y^-_k \sim \mathcal{N}\left(H_k m^-_k, \, H_k P^-_k H_k^T + \Sigma_m\right)$$

---

## 3. Joint Distribution of $\begin{bmatrix} x^-_k \\ y^-_k \end{bmatrix}$

Since both $x^-_k$ and $y^-_k$ are linear transformations of underlying independent Gaussian noises, they are jointly Gaussian:


$$\begin{bmatrix} x^-_k \\ y^-_k \end{bmatrix} \sim \mathcal{N}\left( \begin{bmatrix} \mathbb{E}[x^-_k] \\ \mathbb{E}[y^-_k] \end{bmatrix}, \begin{bmatrix} \text{Var}(x^-_k) & \text{Cov}(x^-_k, y^-_k) \\ \text{Cov}(y^-_k, x^-_k) & \text{Var}(y^-_k) \end{bmatrix} \right)$$

From parts 1 and 2, we already have the means and diagonal covariance blocks. Let's find the cross-covariance $\text{Cov}(x^-_k, y^-_k)$:


$$\text{Cov}(x^-_k, y^-_k) = \mathbb{E}\left[(x^-_k - m^-_k)(y^-_k - \mathbb{E}[y^-_k])^T\right]$$

Substitute $y^-_k - \mathbb{E}[y^-_k] = H_k(x^-_k - m^-_k) + z_k$:


$$\text{Cov}(x^-_k, y^-_k) = \mathbb{E}\left[(x^-_k - m^-_k)\left(H_k(x^-_k - m^-_k) + z_k\right)^T\right]$$

$$\text{Cov}(x^-_k, y^-_k) = \mathbb{E}\left[(x^-_k - m^-_k)(x^-_k - m^-_k)^T H_k^T\right] + \mathbb{E}\left[(x^-_k - m^-_k)z_k^T\right]$$

Because the state prediction error $(x^-_k - m^-_k)$ depends only on past process noises and is totally independent of current measurement noise $z_k$, the second term is zero:


$$\text{Cov}(x^-_k, y^-_k) = \text{Var}(x^-_k)H_k^T = P^-_k H_k^T$$

Symmetrically, $\text{Cov}(y^-_k, x^-_k) = H_k P^-_k$. Putting it all together:


$$\begin{bmatrix} x^-_k \\ y^-_k \end{bmatrix} \sim \mathcal{N}\left( \begin{bmatrix} m^-_k \\ H_k m^-_k \end{bmatrix}, \begin{bmatrix} P^-_k & P^-_k H_k^T \\ H_k P^-_k & H_k P^-_k H_k^T + \Sigma_m \end{bmatrix} \right)$$

---

## 4 & 5. Conditional Distribution (The Kalman Update)

The conditional distribution $x^+_k \triangleq (x^-_k \mid y^-_k = y^\text{obs}_k)$ relies on the standard **Gaussian Conditioning Theorem**.

> **Gaussian Conditioning Lemma:** If $\begin{bmatrix} X \\ Y \end{bmatrix} \sim \mathcal{N}\left(\begin{bmatrix} \mu_x \\ \mu_y \end{bmatrix}, \begin{bmatrix} \Sigma_{xx} & \Sigma_{xy} \\ \Sigma_{yx} & \Sigma_{yy} \end{bmatrix}\right)$, then the conditional distribution of $X$ given $Y=y$ is:
>
> $$\mathbb{E}[X \mid Y=y] = \mu_x + \Sigma_{xy}\Sigma_{yy}^{-1}(y - \mu_y)$$
>
>
> $$\text{Var}(X \mid Y=y) = \Sigma_{xx} - \Sigma_{xy}\Sigma_{yy}^{-1}\Sigma_{yx}$$
>
>

Applying this directly to our joint matrix from part 3:

* $\mu_x = m^-_k$
* $\mu_y = H_k m^-_k$
* $\Sigma_{xx} = P^-_k$
* $\Sigma_{xy} = P^-_k H_k^T$
* $\Sigma_{yy} = H_k P^-_k H_k^T + \Sigma_m$

### Defining the Kalman Gain ($K_k$)

Let $K_k = \Sigma_{xy}\Sigma_{yy}^{-1}$:


$$K_k = P^-_k H_k^T \left(H_k P^-_k H_k^T + \Sigma_m\right)^{-1}$$

### Finding the Conditional Mean (Question 5 part 1)

$$\mathbb{E}[x^-_k \mid y^-_k = y^\text{obs}_k] = m_k = m^-_k + K_k \left(y^\text{obs}_k - H_k m^-_k\right)$$

### Finding the Conditional Variance (Question 5 part 2)

$$\text{Var}(x^-_k \mid y^-_k = y^\text{obs}_k) = P_k = P^-_k - K_k \Sigma_{yx}$$


Substitute $\Sigma_{yx} = H_k P^-_k$:


$$P_k = P^-_k - K_k H_k P^-_k = (I - K_k H_k)P^-_k$$

Therefore, $(x^-_k \mid y^-_k = y^\text{obs}_k) \sim \mathcal{N}(m_k, P_k)$ with the exact update equations shown above.

# Q. 1-D Example

Consider the scalar linear-Gaussian filter model
\begin{aligned}
x^-_k &= a\,x^+_{k-1} + w_{k-1},\qquad w_{k-1}\sim\mathscr N(0,q),\\
y^-_k &= h\,x^-_k + z_k,\qquad\;\;\;\; z_k\sim\mathscr N(0,r),
\end{aligned}
with prior $x_0\sim \mathscr{N}(m_0,P_0)$. Define $Y_{k}=\{y^{\mathrm{obs}}_1,\dots,y^{\mathrm{obs}}_{k}\}$.

Answer the following:

1. Show that
\begin{aligned}
m_k^- &= a\,m_{k-1},\\
P_k^- &= a^2 P_{k-1} + q.
\end{aligned}

2. Show that
\begin{aligned}
m_k &= m_k^- + K_k\,v_k
= m_k^- + \frac{P_k^- h}{S_k}\bigl(y^{\mathrm{obs}}_k - h\,m_k^-\bigr),\\
P_k &= (1 - K_k h)\,P_k^-
= \Bigl(1 - \frac{P_k^- h^2}{S_k}\Bigr) P_k^- .
\end{aligned}

3. Show that the predictive measurement distribution (before seeing $y_k$):
\begin{align*}
p(y^-_k | Y_{k-1})=\mathscr N\bigl(h\,m_k^-,\; h^2 P_k^- + r\bigr).
\end{align*}

4. Show that the posterior-predictive measurement distribution (after filtering on $y_k$):
\begin{align*}
p(y^-_k\mid Y_k)=\mathscr N\bigl(h\,m_k,\; h^2 P_k + r\bigr).
\end{align*}

5. For suitable numerical values for $a,q,h,r$ animate the above to prior distribution and the posterior distribution.

##Q2 Answers

1. State Prediction Step
We are given the dynamic model:
$$x^-_k = a x^+_{k-1} + w_{k-1}, \quad w_{k-1} \sim \mathcal{N}(0, q)$$
Given that $x^+_{k-1} \sim \mathcal{N}(m_{k-1}, P_{k-1})$ and $w_{k-1}$ is independent of $x^+_{k-1}$, the predicted state $x^-_k$ is a linear combination of independent Gaussian variables, so it is also Gaussian: $x^-_k \sim \mathcal{N}(m^-_k, P^-_k)$.

Expected Value (Mean $m^-_k$)
Using the linearity of expectation:
$$m^-_k = \mathbb{E}[x^-_k] = \mathbb{E}[a x^+_{k-1} + w_{k-1}] = a \mathbb{E}[x^+_{k-1}] + \mathbb{E}[w_{k-1}]$$
Since $\mathbb{E}[x^+_{k-1}] = m_{k-1}$ and $\mathbb{E}[w_{k-1}] = 0$:
$$m^-_k = a m_{k-1}$$

Variance ($P^-_k$)
Using the variance property $\text{Var}(bX + cY) = b^2\text{Var}(X) + c^2\text{Var}(Y)$ for independent variables:
$$P^-_k = \text{Var}(x^-_k) = \text{Var}(a x^+_{k-1} + w_{k-1}) = a^2 \text{Var}(x^+_{k-1}) + \text{Var}(w_{k-1})$$
Since $\text{Var}(x^+_{k-1}) = P_{k-1}$ and $\text{Var}(w_{k-1}) = q$:
$$P^-_k = a^2 P_{k-1} + q$$

2. Measurement Update Step
We use the standard 1-D Gaussian Conditioning Theorem:
If $\begin{bmatrix} X \\ Y \end{bmatrix} \sim \mathcal{N}\left(\begin{bmatrix} \mu_x \\ \mu_y \end{bmatrix}, \begin{bmatrix} \sigma_x^2 & \sigma_{xy} \\ \sigma_{yx} & \sigma_y^2 \end{bmatrix}\right)$, then $X \mid Y=y \sim \mathcal{N}(m, P)$ where:
$$m = \mu_x + \frac{\sigma_{xy}}{\sigma_y^2}(y - \mu_y), \quad P = \sigma_x^2 - \frac{\sigma_{xy}^2}{\sigma_y^2}$$
For our model, the joint distribution of $\begin{bmatrix} x^-_k \\ y^-_k \end{bmatrix}$ has:
$\mu_x = m^-_k$, $\sigma_x^2 = P^-_k$
$\mu_y = h m^-_k$, $\sigma_y^2 = S_k = h^2 P^-_k + r$
$\sigma_{xy} = \text{Cov}(x^-_k, h x^-_k + z_k) = h P^-_k$

Defining Kalman Gain ($K_k$) and Innovation ($v_k$)
$$S_k = h^2 P^-_k + r$$
$$K_k = \frac{\sigma_{xy}}{\sigma_y^2} = \frac{P^-_k h}{S_k} = P^-_k h (h^2 P^-_k + r)^{-1}$$
$$v_k = y^\text{obs}_k - h m^-_k$$

Updated Mean ($m_k$)
$$m_k = m^-_k + K_k v_k = m^-_k + \frac{P^-_k h}{S_k}(y^\text{obs}_k - h m^-_k)$$

Updated Variance ($P_k$)
$$P_k = P^-_k - K_k \sigma_{yx} = P^-_k - K_k (h P^-_k) = (1 - K_k h)P^-_k$$
Substituting $K_k = \frac{P^-_k h}{S_k}$:
$$P_k = \left(1 - \frac{P^-_k h^2}{S_k}\right) P^-_k$$

3. Predictive Measurement Distribution $p(y^-_k \mid Y_{k-1})$
Before looking at the new measurement $y_k$, the prediction relies purely on the state history up to $k-1$ (represented by $Y_{k-1}$):
$$y^-_k = h x^-_k + z_k$$
Where $(x^-_k \mid Y_{k-1}) \sim \mathcal{N}(m^-_k, P^-_k)$ and $z_k \sim \mathcal{N}(0,r)$.
Mean: $\mathbb{E}[y^-_k \mid Y_{k-1}] = h \mathbb{E}[x^-_k \mid Y_{k-1}] + \mathbb{E}[z_k] = h m^-_k$
Variance: $\text{Var}(y^-_k \mid Y_{k-1}) = h^2 \text{Var}(x^-_k \mid Y_{k-1}) + \text{Var}(z_k) = h^2 P^-_k + r$
$$p(y^-_k \mid Y_{k-1}) = \mathcal{N}(h m^-_k, h^2 P^-_k + r)$$

4. Posterior-Predictive Measurement Distribution $p(y^-_k \mid Y_k)$
After we process and ingest the snapshot measurement at step $k$, our updated understanding of the underlying state becomes $x_k \mid Y_k \sim \mathcal{N}(m_k, P_k)$.
If we hypothetically re-verify or re-observe a measurement validation branch at that exact time step $k$ using our updated estimate:
$$y^-_k = h x_k + z_k$$
Mean: $\mathbb{E}[y^-_k \mid Y_k] = h \mathbb{E}[x_k \mid Y_k] = h m_k$
Variance: $\text{Var}(y^-_k \mid Y_k] = h^2 \text{Var}(x_k \mid Y_k) + \text{Var}(z_k) = h^2 P_k + r$
$$p(y^-_k \mid Y_k) = \mathcal{N}(h m_k, h^2 P_k + r)$$

This concludes the analytical derivations for the 1-D Kalman Filter as requested in questions 1-4.

5. Animation Code (Python Script)

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from scipy.stats import norm

# --- Parameters ---
a = 0.95    # State transition parameter
q = 0.15    # Process noise variance
h = 1.2     # Measurement model scale
r = 0.25    # Measurement noise variance

# Initial conditions
m0, P0 = 0.0, 1.0
np.random.seed(42)

# Generate true trajectory and noisy measurements for 20 steps
steps = 20
x_true = np.zeros(steps)
y_obs = np.zeros(steps)
x_curr = np.random.normal(m0, np.sqrt(P0))

for k in range(steps):
    x_curr = a * x_curr + np.random.normal(0, np.sqrt(q))
    x_true[k] = x_curr
    y_obs[k] = h * x_curr + np.random.normal(0, np.sqrt(r))

# --- Set up Animation plot ---
fig, ax = plt.subplots(figsize=(10, 6))
x_axis = np.linspace(-4, 4, 1000)

# Pre-compute Kalman Filter steps for tracking states
means_prior, vars_prior = [], []
means_post, vars_post = [], []

m_curr, P_curr = m0, P0
for k in range(steps):
    # Predict step
    m_pred = a * m_curr
    P_pred = a**2 * P_curr + q
    means_prior.append(m_pred)
    vars_prior.append(P_pred)

    # Update step
    S = h**2 * P_pred + r
    K = (P_pred * h) / S
    m_curr = m_pred + K * (y_obs[k] - h * m_pred)
    P_curr = (1 - K * h) * P_pred
    means_post.append(m_curr)
    vars_post.append(P_curr)

# Animation updates
def update(frame):
    ax.clear()

    # Extract states for current frame
    mp, vp = means_prior[frame], vars_prior[frame]
    m_post, v_post = means_post[frame], vars_post[frame]

    # Plot curves
    prior_pdf = norm.pdf(x_axis, mp, np.sqrt(vp))
    post_pdf = norm.pdf(x_axis, m_post, np.sqrt(v_post))

    ax.plot(x_axis, prior_pdf, 'r--', label=rf'Prior $p(x_k \mid Y_{{k-1}})$ (\sigma^2={vp:.2f})', linewidth=2)
    ax.fill_between(x_axis, prior_pdf, color='red', alpha=0.1)

    ax.plot(x_axis, post_pdf, 'g-', label=rf'Posterior $p(x_k \mid Y_k)$ (\sigma^2={v_post:.2f})', linewidth=2)
    ax.fill_between(x_axis, post_pdf, color='green', alpha=0.15)

    # Visual cues for measurement context
    ax.axvline(x_true[frame], color='black', linestyle='-', label=f'True Hidden State: {x_true[frame]:.2f}')
    ax.axvline(y_obs[frame]/h, color='blue', linestyle=':', label=rf'Scaled Observation ($y_{{obs}}/h$): {y_obs[frame]/h:.2f}')

    ax.set_title(f'Kalman Filter Dynamic Evolution — Time Step k = {frame+1}', fontsize=12)
    ax.set_xlabel('State value ($x$)', fontsize=10)
    ax.set_ylabel('Probability Density', fontsize=10)
    ax.set_ylim(0, 1.5)
    ax.legend(loc='upper left')
    ax.grid(True, alpha=0.3)

ani = FuncAnimation(fig, update, frames=steps, interval=1200, repeat=True)
# To display the animation in Colab, the animation object
# should be the last expression in the cell.
ani
```

1. Derivation of the State Transition Matrix $A$For a constant-velocity system, the position changes over a time step $\Delta t$ based on the velocity, while the velocity is assumed to remain constant (up to random process noise perturbations).The physics equations for each independent axis ($x$ and $y$) are:$p_x(k) = p_x(k-1) + v_x(k-1)\Delta t$$p_y(k) = p_y(k-1) + v_y(k-1)\Delta t$$v_x(k) = v_x(k-1)$$v_y(k) = v_y(k-1)$We can write this system of linear equations in matrix form, mapping $x^+_{k-1}$ to $x^-_k$:$$\begin{bmatrix} p_x(k) \\ p_y(k) \\ v_x(k) \\ v_y(k) \end{bmatrix} = \begin{bmatrix} 1 & 0 & \Delta t & 0 \\ 0 & 1 & 0 & \Delta t \\ 0 & 0 & 1 & 0 \\ 0 & 0 & 0 & 1 \end{bmatrix} \begin{bmatrix} p_x(k-1) \\ p_y(k-1) \\ v_x(k-1) \\ v_y(k-1) \end{bmatrix}$$Matching this with the deterministic part of the model $x^-_k = A x^+_{k-1}$, we get:$$A = \begin{bmatrix} 1 & 0 & \Delta t & 0 \\ 0 & 1 & 0 & \Delta t \\ 0 & 0 & 1 & 0 \\ 0 & 0 & 0 & 1 \end{bmatrix}$$2. Derivation of the Measurement Matrix $H$The measurement vector $y_k$ contains only the observed position components:$$y_k = \begin{bmatrix} p^{\text{meas}}_x(k) \\ p^{\text{meas}}_y(k) \end{bmatrix}$$We want an extraction matrix $H$ that isolates the position elements from the full state vector $x_k$:$$\begin{bmatrix} p^{\text{meas}}_x(k) \\ p^{\text{meas}}_y(k) \end{bmatrix} = H \begin{bmatrix} p_x(k) \\ p_y(k) \\ v_x(k) \\ v_y(k) \end{bmatrix} + z_k$$To pick out $p_x(k)$ for the first measurement row and $p_y(k)$ for the second measurement row, $H$ must take the form:$$H = \begin{bmatrix} 1 & 0 & 0 & 0 \\ 0 & 1 & 0 & 0 \end{bmatrix}$$3. Derivation of the Noise Matrix $G$In a constant-velocity model, random physical perturbations are typically modeled as random accelerations acting on the system during the interval $\Delta t$. Let $w_{k-1} = \begin{bmatrix} a_x \\ a_y \end{bmatrix}$ be the white noise acceleration vector.According to Newtonian physics, a constant acceleration vector $w_{k-1}$ over an interval $\Delta t$ affects positions and velocities through integration:$\Delta p_x = \frac{1}{2}\Delta t^2 a_x$$\Delta p_y = \frac{1}{2}\Delta t^2 a_y$$\Delta v_x = \Delta t a_x$$\Delta v_y = \Delta t a_y$Putting these acceleration noise impacts into matrix format yields:$$\begin{bmatrix} \Delta p_x \\ \Delta p_y \\ \Delta v_x \\ \Delta v_y \end{bmatrix} = \begin{bmatrix} \frac{1}{2}\Delta t^2 & 0 \\ 0 & \frac{1}{2}\Delta t^2 \\ \Delta t & 0 \\ 0 & \Delta t \end{bmatrix} \begin{bmatrix} a_x \\ a_y \end{bmatrix}$$Matching this with the stochastic process noise term $G w_{k-1}$, we confirm:$$G = \begin{bmatrix} \frac{1}{2}\Delta t^2 & 0 \\ 0 & \frac{1}{2}\Delta t^2 \\ \Delta t & 0 \\ 0 & \Delta t \end{bmatrix}$$

## Part B

Develop a python based method to filter a sequence of noisy GPS obtained position meaurements.

```python
import numpy as np
import matplotlib.pyplot as plt

class GPSKalmanFilter2D:
    def __init__(self, dt, sigma_p, sigma_m):
        """
        Initializes the 2D Constant-Velocity Kalman Filter.

        Parameters:
        dt      : float - Time step interval between measurements
        sigma_p : float - Standard deviation of the process acceleration noise
        sigma_m : float - Standard deviation of the GPS measurement noise
        """
        self.dt = dt

        # State Transition Matrix (A)
        self.A = np.array([
            [1, 0, dt,  0],
            [0, 1,  0, dt],
            [0, 0,  1,  0],
            [0, 0,  0,  1]
        ])

        # Noise Shaping Matrix (G)
        self.G = np.array([
            [0.5 * dt**2, 0],
            [0, 0.5 * dt**2],
            [dt,          0],
            [0,          dt]
        ])

        # Measurement Matrix (H) - extracts positions
        self.H = np.array([
            [1, 0, 0, 0],
            [0, 1, 0, 0]
        ])

        # Process Noise Covariance (Sigma_p mapped through G)
        # Assuming independent x and y acceleration noises
        Q_accent = np.eye(2) * (sigma_p ** 2)
        self.Q = self.G @ Q_accent @ self.G.T

        # Measurement Noise Covariance (Sigma_m)
        self.R = np.eye(2) * (sigma_m ** 2)

        # Initialize state mean and covariance
        self.m = np.zeros((4, 1))
        self.P = np.eye(4) * 10.0 # High initial uncertainty

    def init_state(self, initial_x, initial_y):
        """Warm start the filter with the first known GPS coordinate."""
        self.m = np.array([[initial_x], [initial_y], [0.0], [0.0]])
        self.P = np.eye(4) * 1.0

    def predict(self):
        """Predicts the next state distribution ahead of time."""
        self.m = self.A @ self.m
        self.P = self.A @ self.P @ self.A.T + self.Q

    def update(self, y_obs):
        """Updates the state estimate with a new GPS position measurement."""
        # Innovation / Residual
        v = y_obs - (self.H @ self.m)

        # Innovation Covariance (S)
        S = self.H @ self.P @ self.H.T + self.R

        # Kalman Gain (K)
        K = self.P @ self.H.T @ np.linalg.inv(S)

        # Posterior Update
        self.m = self.m + K @ v
        self.P = (np.eye(4) - K @ self.H) @ self.P

    def filter_sequence(self, measurements):
        """
        Filters a sequence of N coordinates.
        measurements: np.ndarray of shape (N, 2)
        Returns: np.ndarray of shape (N, 4) containing [px, py, vx, vy]
        """
        N = len(measurements)
        filtered_states = np.zeros((N, 4))

        # Initialize with first observation
        self.init_state(measurements[0, 0], measurements[0, 1])
        filtered_states[0] = self.m.flatten()

        for i in range(1, N):
            self.predict()
            y_obs = measurements[i].reshape(2, 1)
            self.update(y_obs)
            filtered_states[i] = self.m.flatten()

        return filtered_states

# --- Example Usage & Simulation Verification ---
if __name__ == "__main__":
    # Define parameters
    total_time = 30 # seconds
    dt = 1.0        # 1 Hz GPS updates
    steps = int(total_time / dt)

    sigma_p = 0.2   # physical maneuvering acceleration jitter
    sigma_m = 3.0   # GPS error radius roughly 3 meters

    # 1. Simulate a true path moving at constant velocity (with slight noise)
    true_vx, true_vy = 2.0, 1.5
    true_path = np.zeros((steps, 2))
    current_pos = np.array([0.0, 0.0])

    for k in range(steps):
        current_pos += np.array([true_vx, true_vy]) * dt + np.random.normal(0, 0.5 * sigma_p * dt**2, 2)
        true_path[k] = current_pos

    # 2. Add high variance GPS noise to generate measurements
    noisy_gps = true_path + np.random.normal(0, sigma_m, size=true_path.shape)

    # 3. Instantiate and run the filter
    kf = GPSKalmanFilter2D(dt, sigma_p, sigma_m)
    filtered_results = kf.filter_sequence(noisy_gps)

    # 4. Visualization
    plt.figure(figsize=(10, 6))
    plt.plot(true_path[:, 0], true_path[:, 1], 'g-', label='True Underlying Path', linewidth=2)
    plt.scatter(noisy_gps[:, 0], noisy_gps[:, 1], color='red', alpha=0.5, s=25, label='Noisy Raw GPS')
    plt.plot(filtered_results[:, 0], filtered_results[:, 1], 'b-x', label='Kalman Filtered Estimate', linewidth=2)

    plt.title('2D Constant-Velocity Kalman Filter on GPS Telemetry')
    plt.xlabel('X Position (meters)')
    plt.ylabel('Y Position (meters)')
    plt.legend()
    plt.grid(True, alpha=0.4)
    plt.show()
```
