# Q. Statistical Mechanics of the Quantum Oscillator

Consider a single quantum harmonic oscillator of angular frequency $\omega$ in thermal equilibrium with a heat bath at temperature $T$. The quantum mechanical energy eigenvalues are given by:

$$E_j = \hbar \omega \left(j + \frac{1}{2}\right), \quad \text{for } j = 0, 1, 2, \dots$$

where $\hbar$ is the reduced Planck constant and $k_B$ is the Boltzmann constant.

**Answer the Following:**

1. **The Partition Function:**
Show that the canonical partition function $Z(T) = \sum_{j=0}^\infty e^{-E_j / (k_B T)}$ forms a convergent geometric series that simplifies to:

$$Z(T) = \frac{e^{-\frac{\hbar \omega}{2k_BT}}}{1 - e^{-\frac{\hbar \omega}{k_BT}}}$$


2. **Thermodynamic Potentials:**
Using your expression for $Z(T)$, derive explicit formulas for:
* The **Helmholtz Free Energy**, $F(T) = -k_B T \ln Z$.
* The **Internal Energy**, $U(T) = -\frac{\partial \ln Z}{\partial \beta}$ (where $\beta = \frac{1}{k_B T}$).


3. **Entropy:**
Find the **Entropy** $S(T)$ of the system using the thermodynamic relation $S = -\left(\frac{\partial F}{\partial T}\right)$, and verify that your result satisfies $U = F + TS$.
4. **Heat Capacity and Hyperbolic Form:**
* Calculate the constant-volume **Heat Capacity**, $C_V = \left(\frac{\partial U}{\partial T}\right)$.
* Use the hyperbolic identity $\sinh(x) = \frac{e^x - e^{-x}}{2}$ to prove that your expression for $C_V$ can be compactly rewritten as:

$$C_V = k_B \left(\frac{\hbar \omega}{2k_BT}\right)^2 \frac{1}{\sinh^2\left(\frac{\hbar \omega}{2k_BT}\right)}$$




5. **Asymptotic Behavior (Discussion):**
Briefly describe the behavior of the heat capacity $C_V$ in the high-temperature limit ($k_B T \gg \hbar \omega$). Does it match the classical equipartition theorem prediction?

### 1. The Partition Function

The canonical partition function $Z(T)$ for a system in thermal equilibrium at temperature $T$ is given by the sum over all possible energy states $j$:

$$Z(T) = \sum_{j=0}^\infty e^{-E_j / (k_B T)}$$

Given the energy eigenvalues for a quantum harmonic oscillator:

$$E_j = \hbar \omega \left(j + \frac{1}{2}\right)$$

Substitute $E_j$ into the partition function formula:

$$Z(T) = \sum_{j=0}^\infty e^{-\frac{\hbar \omega (j + 1/2)}{k_B T}}$$

Let's separate the terms and introduce $\beta = \frac{1}{k_B T}$ for simplicity:

$$Z(T) = \sum_{j=0}^\infty e^{-\beta \hbar \omega (j + 1/2)}$$

$$Z(T) = \sum_{j=0}^\infty e^{-\beta \hbar \omega j} \cdot e^{-\beta \hbar \omega / 2}$$

We can factor out the term that does not depend on $j$:

$$Z(T) = e^{-\beta \hbar \omega / 2} \sum_{j=0}^\infty (e^{-\beta \hbar \omega})^j$$

This is a geometric series of the form $\sum_{j=0}^\infty r^j$, where $r = e^{-\beta \hbar \omega}$.

For a geometric series to converge, we must have $|r| < 1$. Since $\hbar$, $\omega$, $k_B$, and $T$ are all positive, $e^{-\beta \hbar \omega} = e^{-\frac{\hbar \omega}{k_B T}}$ will always be between 0 and 1, so the series converges.

The sum of an infinite convergent geometric series is $\frac{1}{1-r}$. Therefore:

$$Z(T) = e^{-\beta \hbar \omega / 2} \frac{1}{1 - e^{-\beta \hbar \omega}}$$

Substituting $\beta = \frac{1}{k_B T}$ back into the expression:

$$Z(T) = \frac{e^{-\frac{\hbar \omega}{2k_BT}}}{1 - e^{-\frac{\hbar \omega}{k_BT}}}$$

# Q. The Maximum Entropy Distribution under Mean and Variance Constraints

Consider the random variable $X \in (-\infty, \infty)$. Let:

1. **The Mean:** $\mathbb{E}[X] = \mu$
2. **The Variance:** $\mathbb{E}[(X-\mu)^2] = \sigma^2$

Show that the law of $X$ that maximizes entropy is given by:

$$X \sim \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$$

### The Maximum Entropy Distribution under Mean and Variance Constraints

To find the probability distribution $p(x)$ that maximizes entropy subject to given constraints, we use the method of Lagrange multipliers. For a continuous random variable $X$, the differential entropy is given by:

$$H = - \int_{-\infty}^{\infty} p(x) \ln p(x) \ dx$$

We have the following constraints:

1.  **Normalization:** The total probability must be 1.
    $$\int_{-\infty}^{\infty} p(x) \ dx = 1$$

2.  **Mean:** The expected value of $X$ is $\mu$.
    $$\int_{-\infty}^{\infty} x p(x) \ dx = \mu$$

3.  **Second Moment (related to Variance):** The expected value of $X^2$ is $\mathbb{E}[X^2] = \sigma^2 + \mu^2$.
    $$\int_{-\infty}^{\infty} x^2 p(x) \ dx = \sigma^2 + \mu^2$$

    (Note: Using the second moment $\mathbb{E}[X^2]$ instead of $\mathbb{E}[(X-\mu)^2]$ simplifies the algebra slightly, as $\mathbb{E}[(X-\mu)^2] = \mathbb{E}[X^2] - (\mathbb{E}[X])^2 = \mathbb{E}[X^2] - \mu^2$. So, $\mathbb{E}[X^2] = \sigma^2 + \mu^2$ is an equivalent constraint.)


We form the Lagrangian $L$ by adding the constraints multiplied by their respective Lagrange multipliers ($\lambda_0$, $\lambda_1$, $\lambda_2$):

$$L(p(x), \lambda_0, \lambda_1, \lambda_2) = - \int_{-\infty}^{\infty} p(x) \ln p(x) \ dx + \lambda_0 \left( \int_{-\infty}^{\infty} p(x) \ dx - 1 \right) + \lambda_1 \left( \int_{-\infty}^{\infty} x p(x) \ dx - \mu \right) + \lambda_2 \left( \int_{-\infty}^{\infty} x^2 p(x) \ dx - (\sigma^2 + \mu^2) \right)$$

To maximize $H$, we take the functional derivative of $L$ with respect to $p(x)$ and set it to zero:

$$\frac{\delta L}{\delta p(x)} = - (\ln p(x) + 1) + \lambda_0 + \lambda_1 x + \lambda_2 x^2 = 0$$

Solving for $\ln p(x)$:

$$\ln p(x) = \lambda_0 - 1 + \lambda_1 x + \lambda_2 x^2$$

Exponentiating both sides gives $p(x)$:

$$p(x) = e^{\lambda_0 - 1 + \lambda_1 x + \lambda_2 x^2}$$

We can write $A = e^{\lambda_0 - 1}$ as a normalization constant:

$$p(x) = A e^{\lambda_1 x + \lambda_2 x^2}$$

For $p(x)$ to be a valid probability distribution (i.e., integrate to a finite value), the coefficient of $x^2$ ($\lambda_2$) must be negative. Let $\lambda_2 = -\frac{1}{2k}$ for some positive $k$. And let $\lambda_1 = \frac{m}{k}$.

Then the exponent becomes:

$$\lambda_1 x + \lambda_2 x^2 = \frac{m}{k} x - \frac{1}{2k} x^2 = -\frac{1}{2k} (x^2 - 2mx)$$

Complete the square by adding and subtracting $m^2$ inside the parenthesis:

$$- \frac{1}{2k} (x^2 - 2mx + m^2 - m^2) = - \frac{1}{2k} ((x-m)^2 - m^2) = - \frac{(x-m)^2}{2k} + \frac{m^2}{2k}$$

So, $p(x)$ can be written as:

$$p(x) = A e^{\frac{m^2}{2k}} e^{-\frac{(x-m)^2}{2k}}$$

Combining constants, $p(x) = C e^{-\frac{(x-m)^2}{2k}}$. This is the form of a Gaussian distribution. We now need to match its parameters ($m$, $k$, and $C$) to the constraints $\mu$ and $\sigma^2$.

The general form of a Gaussian distribution is:

$$p(x) = \frac{1}{\sqrt{2\pi \sigma^2}} e^{-\frac{(x-\mu)^2}{2\sigma^2}}$$

By comparing the derived form with the standard Gaussian form, we can identify:

*   The parameter $m$ corresponds to the mean $\mu$.
*   The parameter $k$ corresponds to the variance $\sigma^2$.
*   The constant $C$ corresponds to the normalization constant $\frac{1}{\sqrt{2\pi \sigma^2}}$.

Therefore, the distribution that maximizes entropy subject to fixed mean $\mu$ and variance $\sigma^2$ is the Gaussian (Normal) distribution:

$$p(x) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$$

This demonstrates that the Gaussian distribution is the least-biased distribution when only the mean and variance are known, as it makes no further assumptions about the distribution beyond what is specified by these constraints.

# Q. Multiclass Logistic Regression: Automatic Classification of Emails

An e-commerce platform automatically routes inbound customer support emails to one of four mutually exclusive target destinations ($Y$):

* $Y = 0$ : **Finance**
* $Y = 1$ : **Technical**

Before final classification occurs, an upstream **AI Feature Extractor Agent** processes the raw text of each incoming email and outputs a discrete binary state observation vector $\mathbf{x} = (x_1, x_2, x_3)^T \in \mathscr{X}$, where:

* $x_1 = \mathbb{I}(\text{Financial context detected})$ (e.g., words like *"charge"*, *"invoice"*)
* $x_2 = \mathbb{I}(\text{Account access context detected})$ (e.g., words like *"password"*, *"login"*)
* $x_3 = \mathbb{I}(\text{Bug/malfunction context detected})$ (e.g., words like *"error"*, *"frozen"*)

---

Because human language inherently overlaps, the AI Agent's outputs are distributed non-deterministically across three common operational scenarios. The marginal probability of observing each state is fixed by historical logs to:

1. **Scenario A (Pure Financial):** $\mathbf{x}_A = (1, 0, 0)^T \implies P(\mathbf{x}_A) = 0.40$
2. **Scenario B (Pure Acces):** $\mathbf{x}_B = (0, 1, 0)^T \implies P(\mathbf{x}_B) = 0.25$
3. **Scenario C (Pure Technical):** $\mathbf{x}_C = (0, 0, 1)^T \implies P(\mathbf{x}_C) = 0.05$
3. **Scenario D (Access Technical):** $\mathbf{x}_D = (0, 1, 1)^T \implies P(\mathbf{x}_D) = 0.15$
3. **Scenario E (Billing Access):** $\mathbf{x}_E = (1, 1, 0)^T \implies P(\mathbf{x}_E) = 0.05$

---

Historical manual audits provide the following observed joint dataset expectations:

* Constraint 1: $\mathbb{E}[x_1 \cdot \mathbb{I}(Y=1)] = \mathbf{0.05}$
(Very low. If financial words are present, it is rarely routed to Technical support.)

* Constraint 2: $\mathbb{E}[x_2 \cdot \mathbb{I}(Y=1)] = \mathbf{0.15}$
(Moderate. Account access issues split down the middle depending on context.)

* Constraint 3: $\mathbb{E}[x_3 \cdot \mathbb{I}(Y=1)] = \mathbf{0.16}$
(High. If a bug or malfunction context is detected, it goes to the Tech team the vast majority of the time, utilizing $80\%$ of the $0.20$ max capacity).




### Approach to Multiclass Logistic Regression with Given Constraints

This problem asks us to build a classifier that routes customer support emails to one of four mutually exclusive target destinations ($Y$). However, the problem statement provided only defines two explicit `Y` values ($Y=0$: Finance, $Y=1$: Technical). It is crucial to clarify if the other two destinations are implicitly handled by $Y=0$ or $Y=1$, or if they represent separate classes not yet fully described in the problem statement.

Assuming, for the scope of this explanation, that the problem aims to model the probability of an email going to the 'Technical' department ($Y=1$) versus 'Finance' ($Y=0$) as the primary classification task (a binary logistic regression), we can proceed as follows:

**1. Define the Model:**

For binary logistic regression, we model the probability of $Y=1$ (Technical) given the feature vector $\mathbf{x} = (x_1, x_2, x_3)^T$ using the sigmoid function:

$$P(Y=1 | \mathbf{x}; \mathbf{w}) = \frac{1}{1 + e^{-(\mathbf{w}^T \mathbf{x} + b)}}$$

where $\mathbf{w} = (w_1, w_2, w_3)^T$ are the weights corresponding to each feature, and $b$ is the bias term. For simplicity, we can augment $\mathbf{x}$ with a $1$ and include $b$ in $\mathbf{w}$ to get $P(Y=1 | \mathbf{x}; \mathbf{w}) = \frac{1}{1 + e^{-\mathbf{w}^T \mathbf{x}}}$.

The probability of $Y=0$ (Finance) is then $P(Y=0 | \mathbf{x}; \mathbf{w}) = 1 - P(Y=1 | \mathbf{x}; \mathbf{w})$.

**2. Incorporate Constraints using Maximum Entropy Principle:**

The given constraints are in the form of expected values:

*   $\mathbb{E}[x_1 \cdot \mathbb{I}(Y=1)] = 0.05$
*   $\mathbb{E}[x_2 \cdot \mathbb{I}(Y=1)] = 0.15$
*   $\mathbb{E}[x_3 \cdot \mathbb{I}(Y=1)] = 0.16$

These are empirical expectations that our model should ideally match. The Maximum Entropy principle states that, given a set of observed constraints on a probability distribution, the least biased distribution is the one that maximizes entropy while satisfying those constraints. Logistic regression models (and other Generalized Linear Models) can be derived from the Maximum Entropy principle when features are sufficient statistics for the exponential family distribution. In essence, we seek to find the weights $\mathbf{w}$ that make our model's expectations match the empirical expectations.

Specifically, the expected value of a feature $f_k(\mathbf{x}, Y) = x_k \cdot \mathbb{I}(Y=1)$ under our model should be equal to the observed expectation:

$$\sum_{\mathbf{x} \in \mathscr{X}} P(\mathbf{x}) P(Y=1 | \mathbf{x}; \mathbf{w}) x_k = \mathbb{E}[x_k \cdot \mathbb{I}(Y=1)]_{empirical}$$

where $P(\mathbf{x})$ are the marginal probabilities of observing each feature vector (Scenario A, B, C, D, E).

**3. Formulate the Optimization Problem:**

We typically find the parameters $\mathbf{w}$ by maximizing the likelihood of the training data. Given the expected value constraints, this often translates to finding parameters that ensure these expectations are met. This can be viewed as training a logistic regression model where the loss function implicitly (or explicitly, through methods like Lagrange multipliers) tries to satisfy these moment constraints.

The log-likelihood for a dataset with feature vectors $\mathbf{x}_i$ and true labels $y_i$ is:

$$L(\mathbf{w}) = \sum_{i=1}^N \left[ y_i \ln P(Y=1 | \mathbf{x}_i; \mathbf{w}) + (1 - y_i) \ln P(Y=0 | \mathbf{x}_i; \mathbf{w}) \right]$$

However, in this problem, we are given *joint expectations* rather than a dataset of individual (x,y) pairs. This implies we need to find $\mathbf{w}$ such that the model's expected feature-label co-occurrence matches the given empirical values. This is directly related to the principle of Maximum Entropy, where the model parameters are chosen to minimize the Kullback-Leibler (KL) divergence between the empirical distribution (which satisfies the constraints) and the model distribution. The objective function often involves minimizing a cost function that includes these constraints.

**Example for Constraint 1:**

The model's expected value for $x_1 \cdot \mathbb{I}(Y=1)$ would be:

$$\mathbb{E}_{model}[x_1 \cdot \mathbb{I}(Y=1)] = P(\mathbf{x}_A) \cdot P(Y=1 | \mathbf{x}_A; \mathbf{w}) \cdot x_{1,A} + P(\mathbf{x}_E) \cdot P(Y=1 | \mathbf{x}_E; \mathbf{w}) \cdot x_{1,E}$$

We would set this equal to $0.05$ and solve for $\mathbf{w}$. Similar equations would be set up for the other two constraints.

Solving for $\mathbf{w}$ would involve numerical optimization techniques. Typically, one would define a loss function (e.g., cross-entropy loss) that incorporates the observed marginal probabilities $P(\mathbf{x})$ and the target expectations. Gradient descent or similar algorithms could then be used to find the optimal $\mathbf{w}$ and $b$.

**4. Prediction:**

Once the optimal weights $\mathbf{w}$ are learned, for a new incoming email with features $\mathbf{x}$, we would calculate $P(Y=1 | \mathbf{x}; \mathbf{w})$. If this probability exceeds a certain threshold (e.g., 0.5), it is classified as 'Technical' ($Y=1$), otherwise 'Finance' ($Y=0$).

**Further Considerations:**

*   The problem states four mutually exclusive target destinations, but only two are explicitly listed. If there are truly four distinct classes, this would become a true multiclass logistic regression problem, often solved using a 'one-vs-rest' or softmax approach.
*   The marginal probabilities for scenarios A, B, C, D, E sum up to $0.40 + 0.25 + 0.05 + 0.15 + 0.05 = 0.90$. This indicates that there are other scenarios (with a total probability of $0.10$) that are not explicitly described, or these are just the most common ones. This missing information might impact the accuracy of the model if not accounted for.

# Q. Asset Pricing and Portfolio Optimization.



In standard finance, models often assume asset returns follow a perfect Normal (Gaussian) distribution. However, real-world markets experience extreme events, leading to "fat tails" and skewed returns. MaxEnt allows an analyst to infer the **least-biased probability distribution** of an asset's future price based *only* on what the market currently tells us, without forcing it into a rigid Gaussian shape.

**Background:**

You are a quantitative portfolio manager evaluating a highly volatile tech stock, **Asset X**, currently trading at $\$100$. You want to determine the risk-neutral probability distribution of its price, $S_T$, in exactly one month. This distribution will allow you to correctly price exotic derivatives and optimize your portfolio's risk exposure.

**Available Market Data (The Constraints):**

You do not know the true distribution, but you can observe the following macroeconomic aggregates and market-clearing prices from the options exchange:

1. **The Risk-Free Growth:** Based on the risk-free interest rate, the expected future price of the asset must match the forward price:

$$\mathbb{E}[S_T] = \$102$$


2. **The Market Volatility:** Based on standard historical variance, the expected squared price is constrained by:

$$\mathbb{E}[S_T^2] = 10,600$$


3. **Asymmetric Market Sentiment (The "Skew"):** Due to heavy trading on out-of-the-money put options, the market is pricing in a strong downside risk. The expected value of a specific logarithmic penalty (which captures downward crash risk) is constrained to:

$$\mathbb{E}[\ln(S_T)] = 4.60$$



**Objective:**

Formulate and solve the Maximum Entropy problem to find the continuous probability density function $p(S_T)$ for $S_T > 0$.



To find the probability density function p(S_T) that maximizes entropy subject to your specified market constraints, we utilize the method of Lagrange Multipliers within the framework of Maximum Entropy (MaxEnt) distributions.

### 1. The Optimization Problem
We seek the density $p(S_T)$ that maximizes the Shannon entropy:

$$H = - \int_{0}^{\infty} p(S_T) \ln p(S_T) \ dS_T$$

Subject to the following constraints:

*   **Normalization:** $$\int_{0}^{\infty} p(S_T) \ dS_T = 1$$
*   **Risk-Free Growth (Expectation):** $$\int_{0}^{\infty} S_T p(S_T) \ dS_T = 102$$
*   **Market Volatility (Second Moment):** $$\int_{0}^{\infty} S_T^2 p(S_T) \ dS_T = 10,600$$
*   **Asymmetric Sentiment (Logarithmic Penalty):** $$\int_{0}^{\infty} \ln(S_T) p(S_T) \ dS_T = 4.60$$

### 2. Formulating the Lagrangian
Using multipliers $\lambda_0$ (for normalization), $\lambda_1$, $\lambda_2$, and $\lambda_3$, we define the Lagrangian $\mathcal{L}$:

$$\mathcal{L} = - \int_{0}^{\infty} p(S_T) \ln p(S_T) \ dS_T + \lambda_0 \left( \int_{0}^{\infty} p(S_T) \ dS_T - 1 \right) + \lambda_1 \left( \int_{0}^{\infty} S_T p(S_T) \ dS_T - 102 \right) + \lambda_2 \left( \int_{0}^{\infty} S_T^2 p(S_T) \ dS_T - 10600 \right) + \lambda_3 \left( \int_{0}^{\infty} \ln(S_T) p(S_T) \ dS_T - 4.60 \right)$$

Taking the functional derivative with respect to $p(S_T)$ and setting it to zero yields the general form:

$$p(S_T) = e^{\lambda_0 - 1 + \lambda_1 S_T + \lambda_2 S_T^2 + \lambda_3 \ln(S_T)}$$

Simplifying this, we arrive at the Generalized Exponential Distribution:

$$p(S_T) = C S_T^{\lambda_3} e^{\lambda_1 S_T + \lambda_2 S_T^2}$$

(Where $C = e^{-(1+\lambda_0)}$ is the normalization constant.)

### 3. Analysis of the Distribution
This resulting distribution is a variation of the Generalized Gamma distribution, which inherently accounts for the properties observed in your market data:

*   **The Power Term ($S_T^{\lambda_3}$):** By setting $\lambda_3 > 0$, we introduce a "heavy tail" effect as $S_T \to 0$. This mathematically captures the market’s demand for downside protection (the "skew") that a simple Gaussian distribution would ignore.
*   **The Quadratic Term ($e^{\lambda_2 S_T^2}$):** This ensures the distribution decays rapidly for high values of $S_T$, ensuring the integral converges and respects the volatility constraint ($\mathbb{E}[S_T^2]$). Note that $\lambda_2$ must be negative for convergence.
*   **The Linear Term ($e^{\lambda_1 S_T}$):** This balances the mean towards the forward price of 102.

### 4. Solving for Parameters
To find the specific values for $\lambda_1, \lambda_2$, and $\lambda_3$, you must solve the following system of coupled integral equations (obtained by substituting $p(S_T)$ back into the constraints):

$$\int_{0}^{\infty} C S_T^{\lambda_3} e^{\lambda_1 S_T + \lambda_2 S_T^2} \ dS_T = 1$$
$$\int_{0}^{\infty} S_T \left( C S_T^{\lambda_3} e^{\lambda_1 S_T + \lambda_2 S_T^2} \right) \ dS_T = 102$$
$$\int_{0}^{\infty} S_T^2 \left( C S_T^{\lambda_3} e^{\lambda_1 S_T + \lambda_2 S_T^2} \right) \ dS_T = 10600$$
$$\int_{0}^{\infty} \ln(S_T) \left( C S_T^{\lambda_3} e^{\lambda_1 S_T + \lambda_2 S_T^2} \right) \ dS_T = 4.60$$

These integrals are generally solved using numerical integration (e.g., Gaussian Quadrature) and iterative root-finding algorithms (e.g., Newton-Raphson). Given the constraints provided, the system will yield a distribution that is non-symmetric and possesses a "fat" left tail, correctly reflecting the "crash risk" priced into the put options of Asset X.

This MaxEnt approach provides the most robust estimate for derivative pricing, as it incorporates the essential information (mean, variance, and skew) without imposing unverified assumptions about the underlying distribution's shape.
