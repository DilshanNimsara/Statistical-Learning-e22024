### 1. Derivation of Bias in the MLE Covariance Matrix

We begin with the Maximum Likelihood Estimator (MLE) for the population covariance matrix $\Sigma$:

$$\hat{\Sigma}_{\text{MLE}} = \frac{1}{n} \sum_{i=1}^{n} (X_i - \hat{\mu}_n)(X_i - \hat{\mu}_n)^T$$

where $\hat{\mu}_n = \frac{1}{n} \sum_{i=1}^{n} X_i$ represents the empirical sample mean. To evaluate the bias of this estimator, we expand the deviations around the true, latent population mean $\mu$:

$$X_i - \hat{\mu}_n = (X_i - \mu) - (\hat{\mu}_n - \mu)$$

Substituting this identity back into the summation yields:

$$\sum_{i=1}^{n} (X_i - \hat{\mu}_n)(X_i - \hat{\mu}_n)^T = \sum_{i=1}^{n} \left[ (X_i - \mu) - (\hat{\mu}_n - \mu) \right] \left[ (X_i - \mu) - (\hat{\mu}_n - \mu) \right]^T$$

Expanding the product matrix term-by-term:

$$= \sum_{i=1}^{n} (X_i - \mu)(X_i - \mu)^T - \sum_{i=1}^{n} (X_i - \mu)(\hat{\mu}_n - \mu)^T - \sum_{i=1}^{n} (\hat{\mu}_n - \mu)(X_i - \mu)^T + \sum_{i=1}^{n} (\hat{\mu}_n - \mu)(\hat{\mu}_n - \mu)^T$$

Using the identity $\sum_{i=1}^{n} (X_i - \mu) = n(\hat{\mu}_n - \mu)$, the two middle cross-product terms contract and simplify against the final term:

$$= \sum_{i=1}^{n} (X_i - \mu)(X_i - \mu)^T - n(\hat{\mu}_n - \mu)(\hat{\mu}_n - \mu)^T$$

Now, we apply the expectation operator $\mathbb{E}[\cdot]$ to both sides:

$$\mathbb{E}[\hat{\Sigma}_{\text{MLE}}] = \frac{1}{n} \sum_{i=1}^{n} \mathbb{E}\left[(X_i - \mu)(X_i - \mu)^T\right] - \mathbb{E}\left[(\hat{\mu}_n - \mu)(\hat{\mu}_n - \mu)^T\right]$$

By definition of the underlying structural population moments:

* $\mathbb{E}\left[(X_i - \mu)(X_i - \mu)^T\right] = \Sigma$
* $\mathbb{E}\left[(\hat{\mu}_n - \mu)(\hat{\mu}_n - \mu)^T\right] = \text{Var}(\hat{\mu}_n) = \frac{1}{n}\Sigma$

Substituting these moments back into the expectation equation gives:

$$\mathbb{E}[\hat{\Sigma}_{\text{MLE}}] = \frac{1}{n} (n\Sigma) - \frac{1}{n}\Sigma = \Sigma - \frac{1}{n}\Sigma = \left( \frac{n - 1}{n} \right) \Sigma$$

> **Conclusion on Bias:** Because $\mathbb{E}[\hat{\Sigma}_{\text{MLE}}] \neq \Sigma$, the raw MLE estimator is systematically downward biased in finite samples. This mathematical underestimation happens because calculating variations against the *sample mean* rather than the *true population mean* artificially contracts the apparent spatial scattering of the system.

---

### 2. Elimination of Bias via Bessel's Correction

To construct an unbiased covariance estimator ($S$), we scale the maximum likelihood matrix by the reciprocal of the calculated bias factor, namely $\frac{n}{n - 1}$:

$$S = \frac{n}{n - 1} \hat{\Sigma}_{\text{MLE}} = \frac{n}{n - 1} \left( \frac{1}{n} \sum_{i=1}^{n} (X_i - \hat{\mu}_n)(X_i - \hat{\mu}_n)^T \right) = \frac{1}{n - 1} \sum_{i=1}^{n} (X_i - \hat{\mu}_n)(X_i - \hat{\mu}_n)^T$$

Evaluating the expectation confirms that the bias has been successfully eliminated:

$$\mathbb{E}[S] = \frac{n}{n - 1} \mathbb{E}[\hat{\Sigma}_{\text{MLE}}] = \frac{n}{n - 1} \left( \frac{n - 1}{n} \right) \Sigma = \Sigma$$

## Part B: Theoretical Extension (Asymptotic Distributions & Slutsky's Theorem)

### 1. Rigorous Statement of Slutsky's Theorem

Let $\{X_n\}$ and $\{Y_n\}$ be sequences of random variables (or vectors). If $X_n$ converges in distribution to a random variable $X$ ($X_n \xrightarrow{d} X$), and $Y_n$ converges in probability to a constant scalar or matrix $c$ ($Y_n \xrightarrow{p} c$), then the following asymptotic relationships hold:

* **Additivity:**
$$X_n + Y_n \xrightarrow{d} X + c$$


* **Multiplicativity:**
$$X_n Y_n \xrightarrow{d} cX$$


* **Divisibility:**
$$X_n / Y_n \xrightarrow{d} X / c \quad (\text{provided } c \neq 0)$$



---

### 2. Operational Distribution Justification

To validate the parametric modeling framework used in real-time structural health monitoring dashboards, we apply Slutsky's theorem to substitute the sample covariance matrix $S$ for the unobservable population covariance matrix $\Sigma$.

#### Step 1: The Baseline Asymptotic Space

The Lindeberg-Lévy Multivariate Central Limit Theorem (CLT) asserts that as the sample snapshot size approaches infinity ($n \to \infty$), the standardized difference between the empirical mean and population mean converges in distribution to a Multivariate Gaussian:

$$\sqrt{n}(\hat{\mu}_n - \mu) \xrightarrow{d} \mathcal{N}(0, \Sigma)$$

#### Step 2: Normalization of the Vector Space

Multiplying by the inverse matrix square root of the latent population covariance matrix stabilizes and normalizes the variance space to a standard identity profile:

$$\Sigma^{-1/2} \left[ \sqrt{n}(\hat{\mu}_n - \mu) \right] \xrightarrow{d} \mathcal{N}(0, I)$$

#### Step 3: Convergence of the Sample Estimator

By the Weak Law of Large Numbers (WLLN), the unbiased empirical sample covariance matrix $S$ is a continuous, consistent estimator of the true population matrix $\Sigma$. Therefore, as $n \to \infty$, $S$ converges in probability to the constant population profile:

$$S \xrightarrow{p} \Sigma \implies S^{-1/2} \xrightarrow{p} \Sigma^{-1/2}$$

#### Step 4: Application of Slutsky's Theorem

We now multiply the sequence converging in distribution (Step 2) by the sequence converging in probability to a constant matrix (Step 3). Slutsky’s theorem guarantees that the product preserves convergence to the target distribution:

$$S^{-1/2} \left[ \sqrt{n}(\hat{\mu}_n - \mu) \right] \xrightarrow{d} \Sigma^{-1/2} \cdot \mathcal{N}(0, \Sigma) = \mathcal{N}(0, I)$$

#### Step 5: Scaling Back to the Empirical Metric

Because the target boundary converges asymptotically to a standard Gaussian space, we can map the localized coordinate parameters directly back to our empirical matrix metric:

$$\sqrt{n}(\hat{\mu}_n - \mu) \sim \mathcal{N}(0, S) \implies \hat{\mu}_n \sim \mathcal{N}\left(\mu, \frac{1}{n}S\right)$$

> **Conclusion:** This mathematical progression validates substituting $S$ for $\Sigma$ in high-volume environments. It proves that using the sample covariance matrix does not corrupt or compromise the statistical stability of the asymptotic diagnostic envelope.

````python
import numpy as np
import pandas as pd
import scipy.stats as stats

# --- DO NOT MODIFY: Synthetic Strain Sensor Data Generator ---
np.random.seed(42)
n_samples = 5000
n_features = 3

# Generating base stationary multivariate normal noise
base_data = np.random.multivariate_normal(
    mean=[0.5, -0.2, 1.1],
    cov=[[0.09, 0.02, 0.01],
         [0.02, 0.06, 0.03],
         [0.01, 0.03, 0.05]],
    size=n_samples
)

# Injecting a subtle, hidden structural drift in the final 1,000 snapshots
base_data[4000:, 0] += 0.015   # Sensor 1 drift
base_data[4000:, 2] -= 0.010   # Sensor 3 drift

df_strain = pd.DataFrame(base_data, columns=['Strain_Ch1', 'Strain_Ch2', 'Strain_Ch3'])
# ----------------------------------------------------------------

def verify_first_moment_homogeneity(df: pd.DataFrame, g_chunks: int = 5) -> dict:
    """
    Partitions the dataset into g chunks and evaluates first-moment homogeneity
    via Wilks' Lambda and Bartlett's Chi-Square asymptotic transformation.
    """
    X = df.to_numpy()
    n, m = X.shape
    n_j = n // g_chunks  # Size of each chunk (1000)

    # 1. Compute the global mean
    global_mean = np.mean(X, axis=0)

    # Initialize Within-Chunk (W) and Between-Chunk (B) matrices
    W = np.zeros((m, m))
    B = np.zeros((m, m))

    # 2. Partition and compute matrices
    for j in range(g_chunks):
        chunk = X[j * n_j : (j + 1) * n_j]
        chunk_mean = np.mean(chunk, axis=0)

        # Within-Chunk Scatter: Sum of squared deviations within chunks
        dev_within = chunk - chunk_mean
        W += dev_within.T @ dev_within

        # Between-Chunk Scatter: Weighted deviations of chunk means from global mean
        dev_between = (chunk_mean - global_mean).reshape(-1, 1)
        B += n_j * (dev_between @ dev_between.T)

    # 3. Compute Wilks' Lambda (Λ)
    wilks_lambda = np.linalg.det(W) / np.linalg.det(B + W)

    # 4. Transform using Bartlett's Chi-Square approximation
    df_chi = m * (g_chunks - 1)
    chi2_calc = -((n - 1) - (m + g_chunks) / 2) * np.log(wilks_lambda)

    # 5. Calculate the p-value
    p_value = 1.0 - stats.chi2.cdf(chi2_calc, df_chi)

    # 6. Formulate conclusion
    alpha = 0.05
    if p_value < alpha:
        conclusion = (f"Reject H0 at alpha={alpha}. The baseline center has significantly "
                      f"shifted over time (First moment is HETEROGENEOUS).")
    else:
        conclusion = (f"Fail to reject H0 at alpha={alpha}. No significant shift detected "
                      f"over time (First moment is HOMONGENEOUS).")

    return {
        "Wilks' Lambda (Λ)": wilks_lambda,
        "Bartlett Chi-Square": chi2_calc,
        "Degrees of Freedom": df_chi,
        "p-value": p_value,
        "Diagnostic Conclusion": conclusion
    }

# Execute the completed function:
results = verify_first_moment_homogeneity(df_strain, g_chunks=5)
for key, val in results.items():
    if isinstance(val, float):
        print(f"{key}: {val:.6f}")
    else:
        print(f"{key}: {val}")
````

## 1. Eigenvector Orthogonal Projections & Covariance Diagonalization

### Step-by-Step Proof

We are given a zero-mean, centered baseline vector $\tilde{X}_i = (X_i - \mu) \in \mathbb{R}^m$. Its symmetric, positive-semidefinite population covariance matrix is defined by:


$$\Sigma = E[\tilde{X}_i \tilde{X}_T^T]$$

Applying spectral decomposition to $\Sigma$ yields:


$$\Sigma = P \Lambda P^T$$

where $P$ is the orthogonal matrix of eigenvectors ($P^T P = P P^T = I$) and $\Lambda$ is the diagonal matrix of eigenvalues. We project $\tilde{X}_i$ into the principal component space using the mapping:


$$Z_i = P^T \tilde{X}_i$$

To find the theoretical covariance matrix of the transformed vector $Z_i$, we expand the expectation operator $E[Z_i Z_i^T]$:

1. **Substitute the projection definition:**

$$E[Z_i Z_i^T] = E\left[(P^T \tilde{X}_i)(P^T \tilde{X}_i)^T\right]$$


2. **Apply the matrix transpose property $(AB)^T = B^T A^T$:**

$$E[Z_i Z_i^T] = E\left[P^T \tilde{X}_i \tilde{X}_i^T P\right]$$


3. **Move the deterministic matrix $P$ outside the expectation operator:**

$$E[Z_i Z_i^T] = P^T E[\tilde{X}_i \tilde{X}_i^T] P$$


4. **Substitute the raw feature covariance matrix definition ($\Sigma = E[\tilde{X}_i \tilde{X}_i^T]$):**

$$E[Z_i Z_i^T] = P^T \Sigma P$$


5. **Substitute the spectral decomposition of $\Sigma$ ($\Sigma = P \Lambda P^T$):**

$$E[Z_i Z_i^T] = P^T (P \Lambda P^T) P$$


6. **Group the terms to exploit orthogonal properties ($P^T P = I$):**

$$E[Z_i Z_i^T] = (P^T P) \Lambda (P^T P) = I \Lambda I = \Lambda$$



$$E[Z_i Z_i^T] = \Lambda \quad \blacksquare$$

### Geometric & Statistical Meaning

* **Statistical Meaning:** Because $\Lambda$ is a strictly diagonal matrix ($\Lambda = \text{diag}(\lambda_1, \lambda_2, \dots, \lambda_m)$), the cross-covariance between any two distinct coordinates $Z_{i,j}$ and $Z_{i,k}$ (where $j \neq f$) is exactly **zero**. This proves that the linear transformation completely de-correlates the original sensor data stream.
* **Geometric Meaning:** The orthogonal matrix $P$ acts as a rigid rotation of the coordinate system. It rotates the original correlated sensor axes into alignment with the orthogonal hyper-ellipsoid axes of the data's spatial scattering, creating a brand new, mutually independent coordinate system.

## 2. Total Variance Conservation & Subspace Ratios

### Formal Proof of Variance Preservation

The total variance of a multivariate system is defined as the sum of the variances of its individual dimensions, which corresponds algebraically to the trace ($\text{tr}$) of its covariance matrix.

Using the cyclic permutation property of the matrix trace—which states that $\text{tr}(ABC) = \text{tr}(CAB) = \text{tr}(BCA)$—we map the raw space trace to the principal space trace:

1. **Start with the trace of the orthogonalized space covariance $\Lambda$:**

$$\text{tr}(\Lambda) = \text{tr}(P^T \Sigma P)$$


2. **Permute the matrix sequence cyclically** by shifting the trailing orthogonal matrix $P$ to the front:

$$\text{tr}(P^T \Sigma P) = \text{tr}(P P^T \Sigma)$$


3. **Apply the identity property of orthogonal matrices ($P P^T = I$):**

$$\text{tr}(P P^T \Sigma) = \text{tr}(I \Sigma) = \text{tr}(\Sigma)$$



Thus, the total variance remains perfectly conserved across the spatial coordinate rotation:


$$\text{tr}(\Sigma) = \text{tr}(\Lambda) = \sum_{j=1}^{m} \lambda_j \quad \blacksquare$$

### Subspace Partition Formulations

When we truncate our space to retain the first $k$ components using the empirical sorted eigenvalues ($\hat{\lambda}_1 \ge \hat{\lambda}_2 \ge \dots \ge \hat{\lambda}_m$), the mathematically rigorous ratios are:

* **Cumulative Explained Variance Ratio ($\Phi(k)$):** Measures the proportion of global structural energy successfully captured by the primary $k$-dimensional loading subspace ($P_k$).

$$\Phi(k) = \frac{\sum_{j=1}^{k} \hat{\lambda}_j}{\sum_{j=1}^{m} \hat{\lambda}_j}$$


* **Residual Unexplained Variance Ratio ($\Psi(k)$):** Measures the remaining fraction of system information discarded into the orthogonal error subspace ($P_{m-k}$).

$$\Psi(k) = 1 - \Phi(k) = \frac{\sum_{j=k+1}^{m} \hat{\lambda}_j}{\sum_{j=1}^{m} \hat{\lambda}_j}$$

## 3. Subspace Decomposition, Orthogonal Energies & Diagnostics

$$e_i = (x_i - \hat{\mu}_n) - \hat{x}_i = \hat{P}_{m-k} z_{i,m-k}$$



### Pythagorean Subspace Energy Proof

We want to prove that:


$$\|x_i - \hat{\mu}_n\|^2 = \|z_{i,k}\|^2 + \|z_{i,m-k}\|^2$$

1. **Express the squared Euclidean norm as a vector dot product:**

$$\|x_i - \hat{\mu}_n\|^2 = (x_i - \hat{\mu}_n)^T (x_i - \hat{\mu}_n)$$


2. **Substitute the partitioned subspace projection vectors:**

$$\|x_i - \hat{\mu}_n\|^2 = \left(\hat{P}_k z_{i,k} + \hat{P}_{m-k} z_{i,m-k}\right)^T \left(\hat{P}_k z_{i,k} + \hat{P}_{m-k} z_{i,m-k}\right)$$


3. **Expand the matrix multiplication term-by-term:**

$$\|x_i - \hat{\mu}_n\|^2 = z_{i,k}^T \hat{P}_k^T \hat{P}_k z_{i,k} + z_{i,k}^T \hat{P}_k^T \hat{P}_{m-k} z_{i,m-k} + z_{i,m-k}^T \hat{P}_{m-k}^T \hat{P}_k z_{i,k} + z_{i,m-k}^T \hat{P}_{m-k}^T \hat{P}_{m-k} z_{i,m-k}$$


4. **Apply subspace constraints and mutual orthogonality properties:**
* Because eigenvectors are normalized, $\hat{P}_k^T \hat{P}_k = I_{k \times k}$ and $\hat{P}_{m-k}^T \hat{P}_{m-k} = I_{(m-k) \times (m-k)}$.
* Because distinct eigenvectors are strictly perpendicular, the cross-subspace projections drop out completely: $\hat{P}_k^T \hat{P}_{m-k} = 0$ and $\hat{P}_{m-k}^T \hat{P}_k = 0$.


5. **Simplify the remaining energy terms:**

$$\|x_i - \hat{\mu}_n\|^2 = z_{i,k}^T I z_{i,k} + 0 + 0 + z_{i,m-k}^T I z_{i,m-k}$$


$$\|x_i - \hat{\mu}_n\|^2 = z_{i,k}^T z_{i,k} + z_{i,m-k}^T z_{i,m-k}$$



Using the definition of the squared norm ($\|v\|^2 = v^T v$), we arrive exactly at the clean energy decomposition:


$$\|x_i - \hat{\mu}_n\|^2 = \|z_{i,k}\|^2 + \|z_{i,m-k}\|^2 \quad \blacksquare$$

### Answers

## 1. The Fundamental Equation of Factor Analysis

### Step-by-Step Proof

We begin with the linear generative structure of the Factor Analysis model for a normalized, zero-mean baseline feature vector $Z_i \in \mathbb{R}^m$:


$$Z_i = \Lambda F_i + \epsilon_i$$

The universal correlation matrix is defined by the expectation operator:


$$R = E[Z_i Z_i^T]$$

To derive the fundamental equation, we substitute the generative equation into the definition of $R$:

1. **Substitute the model definition for $Z_i$:**

$$R = E\left[(\Lambda F_i + \epsilon_i)(\Lambda F_i + \epsilon_i)^T\right]$$


2. **Apply the matrix transpose property $(AB + C)^T = B^T A^T + C^T$:**

$$R = E\left[(\Lambda F_i + \epsilon_i)(F_i^T \Lambda^T + \epsilon_i^T)\right]$$


3. **Expand the matrix multiplication term-by-term using FOIL:**

$$R = E\left[\Lambda F_i F_i^T \Lambda^T + \Lambda F_i \epsilon_i^T + \epsilon_i F_i^T \Lambda^T + \epsilon_i \epsilon_i^T\right]$$


4. **Distribute the linear expectation operator $E[\cdot]$ across the sum:**

$$R = E[\Lambda F_i F_i^T \Lambda^T] + E[\Lambda F_i \epsilon_i^T] + E[\epsilon_i F_i^T \Lambda^T] + E[\epsilon_i \epsilon_i^T]$$


5. **Pull the deterministic factor loading matrix $\Lambda$ outside the expectations:**

$$R = \Lambda E[F_i F_i^T] \Lambda^T + \Lambda E[F_i \epsilon_i^T] + E[\epsilon_i F_i^T] \Lambda^T + E[\epsilon_i \epsilon_i^T]$$


6. **Apply the core statistical assumptions of the generative model:**
* **Orthogonal common factors:** $E[F_i F_i^T] = I_{k \times k}$
* **Diagonalized measurement noise matrix:** $E[\epsilon_i \epsilon_i^T] = \Psi$
* **Independence between latent factors and errors:** $E[\epsilon_i F_i^T] = 0$ (and consequently $E[F_i \epsilon_i^T] = 0$)


7. **Substitute these identities to simplify the expression:**

$$R = \Lambda (I) \Lambda^T + \Lambda (0) + (0) \Lambda^T + \Psi$$


$$R = \Lambda \Lambda^T + \Psi \quad \blacksquare$$



---

### Parameter Definitions & Physical Interpretations

Looking closely at the $j$-th diagonal entry of the decomposition, the total variance of the $j$-th normalized sensor ($R_{jj} = 1$) is split into two components:


$$1 = \sum_{r=1}^{k} \lambda_{jr}^2 + \phi_j^2$$

#### Communality ($h_j^2$)

* **Mathematical Definition:** The sum of squared factor loadings for the $j$-th sensor across all $k$ shared latent factors.

$$h_j^2 = \sum_{r=1}^{k} \lambda_{jr}^2$$


* **Physical Interpretation:** It represents the proportion of a sensor's variance that is driven by the true, underlying global physics of the asset (such as ambient structural load fields or universal thermal fluctuations). A high communality means the sensor is successfully listening to the shared operational modes.

#### Uniqueness ($\phi_j^2$)

* **Mathematical Definition:** The remaining portion of the variance that cannot be explained by the common factors.

$$\phi_j^2 = 1 - h_j^2$$


* **Physical Interpretation:** It captures the localized noise floor unique to that specific hardware channel. In an asset diagnostics layer, a sudden spike in a channel's uniqueness—while global factors remain stationary—strongly flags an isolated instrument fault, calibration loss, or electrical short rather than true structural damage.

---

## 2. Orthogonal Coordinate Rotations & Simple Structure

### Raw PCA Loadings vs. Varimax-Rotated FA Loadings

* **Raw PCA Loading Vectors:** Focus purely on maximizing the global variance explained in descending order ($\lambda_1 > \lambda_2 > \dots$). As a result, the first few principal components are mathematically forced to be a complex, highly mixed combination of *all* sensors, making physical diagnostic mapping messy.
* **Varimax-Rotated FA Loading Vectors:** Loosen this strict hierarchical energy rule. The Varimax criterion maximizes the variance of the *squared* loadings within each column. Geometrically, it rotates the factor axes to align closely with distinct clusters of sensors, driving loadings toward either near-zero or near-one. This isolates distinct physical phenomena (e.g., separating structural bending modes from localized operational motor heat).

### Invariance of Communality under Orthogonal Rotations

Let $T$ be an orthogonal rotation matrix such that $T T^T = T^T T = I$. The rotated factor loading matrix is defined as $\Lambda^* = \Lambda T$.

Substituting this back into the global covariance matrix approximation shows:


$$R \approx \Lambda^* (\Lambda^*)^T + \Psi$$

$$R \approx (\Lambda T)(\Lambda T)^T + \Psi$$

$$R \approx \Lambda (T T^T) \Lambda^T + \Psi$$

Because $T T^T = I$, the rotation drops out perfectly:


$$R \approx \Lambda I \Lambda^T + \Psi = \Lambda \Lambda^T + \Psi$$

Because the global reconstructed matrix $R$ is perfectly preserved, its individual diagonal entries—which dictate the total shared communality ($h_j^2$) of each sensor—remain completely unchanged. Orthogonal rotation merely alters the viewpoint of the coordinate system, redistributing the variance allocation *across* the columns of $\Lambda$ without leaking any variance into or out of the unique noise floor ($\Psi$).

---

## 3. Latent State Reconstruction via Thomson’s Regression Method

### Step 1: Proving the Joint Covariance Expansion

We construct the joint partitioned vector $Y_i = [Z_i^T, F_i^T]^T$. Its joint population covariance matrix is structured as:


$$\Sigma_{YY} = \begin{bmatrix} E[Z_i Z_i^T] & E[Z_i F_i^T] \\ E[F_i Z_i^T] & E[F_i F_i^T] \end{bmatrix}$$

1. We already know $E[Z_i Z_i^T] = R$ and $E[F_i F_i^T] = I_{k \times k}$.
2. We evaluate the cross-covariance block $E[Z_i F_i^T]$ by substituting the generative framework:

$$E[Z_i F_i^T] = E[(\Lambda F_i + \epsilon_i) F_i^T] = E[\Lambda F_i F_i^T + \epsilon_i F_i^T]$$


$$E[Z_i F_i^T] = \Lambda E[F_i F_i^T] + E[\epsilon_i F_i^T]$$



Using the model assumptions ($E[F_i F_i^T] = I$ and $E[\epsilon_i F_i^T] = 0$):


$$E[Z_i F_i^T] = \Lambda (I) + 0 = \Lambda$$

Taking the transpose gives the matching lower block: $E[F_i Z_i^T] = (E[Z_i F_i^T])^T = \Lambda^T$.

Plugging these back into the block architecture yields:


$$\Sigma_{YY} = \begin{bmatrix} R & \Lambda \\ \Lambda^T & I_{k \times k} \end{bmatrix} \quad \blacksquare$$


*(Note: Depending on vector stacking convention, this matches the block structure $\Sigma_{YY} = \begin{bmatrix} R & \Lambda^T \\ \Lambda & I \end{bmatrix}$).*

---

### Step 2: Derivation of Thomson's Estimator Equation

We wish to compute the factor scores using the conditional expectation $f_i = E[F_i | z_i]$. Under the standard projection identity for a zero-mean Multivariate Normal distribution:


$$E[X_2 | x_1] = \mu_2 + \Sigma_{21} \Sigma_{11}^{-1} (x_1 - \mu_1)$$

Mapping our joint partitioned variables to the identity slots:

* $X_2 \longrightarrow F_i$ (with $\mu_2 = 0$)
* $x_1 \longrightarrow z_i$ (with $\mu_1 = 0$)
* $\Sigma_{11} \longrightarrow R$ (the auto-covariance of the observed space)
* $\Sigma_{21} \longrightarrow \Lambda$ (the cross-covariance block between the latent factors and observed spaces)

Substituting these terms directly into the conditional expectation formula gives:


$$f_i = 0 + \Lambda R^{-1} (z_i - 0)$$

$$f_i = \Lambda R^{-1} z_i \quad \blacksquare$$

*(Note: If utilizing transposed notations across matrices, this maps symmetrically to $f_i = \Lambda^T R^{-1} z_i$, matching standard score execution engines).*

````python
import numpy as np
import pandas as pd

# --- DO NOT MODIFY: Asset Data Generator from Part B ---
np.random.seed(42)
n_samples = 2500

# 1. Generate underlying, independent true physical driving forces (Latent Factors)
f1 = np.random.normal(0, 1, n_samples)   # Primary structural mode/load
f2 = np.random.normal(0, 1, n_samples)   # Secondary operational mode

# 2. Construct the Observable Multi-Sensor Array with targeted noise profiles
s1 = 0.85 * f1 + 0.10 * f2 + np.random.normal(0, 0.3, n_samples)
s2 = 0.80 * f1 + 0.15 * f2 + np.random.normal(0, 0.35, n_samples)
s3 = 0.12 * f1 + 0.90 * f2 + np.random.normal(0, 0.25, n_samples)
s4 = 0.02 * f1 + 0.05 * f2 + np.random.normal(0, 1.40, n_samples)

df_asset = pd.DataFrame(
    data=np.vstack([s1, s2, s3, s4]).T,
    columns=['Sensor_1', 'Sensor_2', 'Sensor_3', 'Sensor_4']
)
# --------------------------------------------------------

class PCASubspaceAnalyzer:
    def __init__(self, df: pd.DataFrame):
        self.df = df
        self.n, self.m = df.shape

    def analyze_truncation_boundaries(self) -> pd.DataFrame:
        """
        Standardizes data, performs an exact eigendecomposition,
        and extracts the empirical expected values of T² and Q statistics
        across truncation cutoffs k = 1, 2, 3.
        """
        # 1. Standardize and center the features (using sample std with ddof=1)
        X_raw = self.df.to_numpy()
        mean_vec = np.mean(X_raw, axis=0)
        std_vec = np.std(X_raw, axis=0, ddof=1)

        # Guardrail check against division by zero
        std_vec = np.where(std_vec == 0, 1e-15, std_vec)
        Z = (X_raw - mean_vec) / std_vec

        # 2. Compute the unbiased sample covariance matrix S
        S = np.cov(Z, rowvar=False)

        # 3. Perform Spectral Decomposition
        eigenvalues, eigenvectors = np.linalg.eigh(S)

        # Sort eigenvalues and matching eigenvectors in strictly descending order
        idx = np.argsort(eigenvalues)[::-1]
        lambdas = eigenvalues[idx]
        P = eigenvectors[:, idx]

        # Project standardized scores into the principal component space
        Z_space = Z @ P

        analysis_records = []

        # 4. Loop through every truncation choice k
        for k in [1, 2, 3]:
            # Hotelling's T² sample vector: Sum over the retained components
            # T²_i = \sum_{j=1}^k (z_{i,j}² / \lambda_j)
            t2_samples = np.sum((Z_space[:, :k] ** 2) / lambdas[:k], axis=1)

            # Residual Q statistic (SPE) sample vector: Sum over discarded components
            # Q_i = \sum_{j=k+1}^m z_{i,j}²
            q_samples = np.sum(Z_space[:, k:] ** 2, axis=1)

            # Calculate the empirical means (Expected Values)
            mean_t2 = np.mean(t2_samples)
            mean_q = np.mean(q_samples)

            analysis_records.append({
                "Subspace Size (k)": k,
                "Expected Value E[T²]": mean_t2,
                "Expected Value E[Q]": mean_q
            })

        return pd.DataFrame(analysis_records)

# Execute the complete analysis pipeline block
analyzer = PCASubspaceAnalyzer(df_asset)
summary_df = analyzer.analyze_truncation_boundaries()

# Print formatted output table
print("=" * 60)
print(f"{'PCA SUBSPACE TRUNCATION PROFILE':^60}")
print("=" * 60)
print(summary_df.to_string(index=False, formatters={
    "Expected Value E[T²]": "{:.6f}".format,
    "Expected Value E[Q]": "{:.6f}".format
}))
print("=" * 60)
````

### 2. How Localized Noise "Tricks" the PCA Engine

PCA is fundamentally blind to the *source* of variance; its sole objective function is to find orthogonal directions that maximize global variance ($\text{tr}(\Sigma)$).

### 3. Major Risks in Automated Anomaly Detection Frameworks

Relying *only* on a PCA framework for an automated plant alert layer introduces two severe operational vulnerabilities:

* **Blindness to True Structural Damage (Masking Effect):** Because the top principal components are corrupted and heavily weighted toward tracking the noisy, non-physical variances of broken sensors, the active tracking subspace ($T^2$) becomes desensitized to real, subtle shifts in the asset's true load paths (like an internal fatigue crack). ### Answers
### 1. How Varimax Rotation Relaxes the Strict PCA Hierarchy to Achieve "Simple Structure"

Traditional PCA functions under a strict mathematical optimization constraint where each sequential principal component must maximize the remaining global system variance ($\lambda_1 > \lambda_2 > \dots$). Because of this top-heavy energy requirement, the first few principal components are mathematically forced to absorb mixed, cross-correlated information from **all** channels simultaneously. This blends distinct physical phenomena into complex linear combinations.

The **Varimax rotation** in Factor Analysis completely relaxes this rigid variance hierarchy. It operates on a fixed latent subspace (e.g., $k=2$) and performs an orthogonal coordinate rotation to maximize the *variance of the squared loadings within each factor column*.

$$\text{Maximize } V = \sum_{r=1}^{k} \left[ \frac{1}{m} \sum_{j=1}^{m} \lambda_{jr}^4 - \left( \frac{1}{m} \sum_{j=1}^{m} \lambda_{jr}^2 \right)^2 \right]$$

Geometrically, this rotation pushes the factor loading weights toward the extremes—driving them either very close to **1.0** or very close to **0.0**. This isolates distinct sensor groupings into a **"simple structure"**, allowing each latent factor to capture a discrete, independent source of variance rather than a mixed combination.

---

### 2. Operational Benefits of the Rotated FA Heatmap for Plant Troubleshooting

From a plant operator’s perspective, an unrotated PCA eigenvector matrix is incredibly difficult to debug during a live asset failure because of this structural blending. If a raw PCA matrix shows high, mixed weights across `Sensor_1`, `Sensor_2`, and `Sensor_3` on a single component, an operator cannot easily determine which subsystem is experiencing an anomaly.

In contrast, the **Varimax-rotated Factor Analysis heatmap** maps directly to the underlying physical machinery of the asset:

* **Direct Physical Mapping:** As seen in your dashboard, `Sensor_1` and `Sensor_2` load heavily onto **Factor 1**, while `Sensor_3` loads exclusively onto **Factor 2**.
* **Isolating Independent Modes:** If `Factor 1` represents a global structural loading mode (e.g., symmetric physical bending or tension across the first two channels) and `Factor 2` represents an operational thermal or mechanical boundary unique to the zone around `Sensor_3`, they are completely decoupled.
