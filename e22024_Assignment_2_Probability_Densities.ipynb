```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import beta

def plot_beta_distribution(alpha, beta_val):
    """
    Plots the Beta distribution PDF for given alpha and beta parameters using Matplotlib.
    """
    x = np.linspace(0, 1, 1000)
    y = beta.pdf(x, alpha, beta_val)

    fig, ax = plt.subplots(figsize=(9, 5))

    ax.plot(x, y, linewidth=2, label=f'Beta(α={alpha}, β={beta_val})')
    ax.fill_between(x, y, alpha=0.15)

    ax.set_title(f'Beta Distribution Density Function (α={alpha}, β={beta_val})', fontsize=14)
    ax.set_xlabel('x')
    ax.set_ylabel('PDF(x)')
    ax.set_xlim(0, 1)
    ax.set_ylim(bottom=0)
    ax.legend()
    ax.grid(True, linestyle='--', alpha=0.4)

    plt.tight_layout()
    return fig

# Example usage:
fig = plot_beta_distribution(alpha=2, beta_val=18)
plt.show()
```

# Q(5)

1. Since $X$ represents the installation time in hours, with best-case and worst-case estimates of $4$ and $10$, the support of $X$ is the interval $[4, 10]$.

$$
4 \leq X \leq 10.
$$

2. Given that $X$ follows a Triangular distribution with parameters $a=4$, $b=10$, and $c=6$, its probability density function (PDF) is defined piecewise. Initially, the density is expressed as:


$$f_X(x)= \begin{cases}
\dfrac{2(x-4)}{(10-4)(6-4)}, & 4 \leq x < 6 \\
\dfrac{2}{10-4}, & x=6 \\
\dfrac{2(10-x)}{(10-4)(10-6)}, & 6 < x \leq 10 \\
 0, & \text{otherwise}
 \end{cases}
 $$

 After calculating the denominators, the PDF simplifies to the following more manageable form:

 $$f_X(x)= \begin{cases}
 \frac{x-4}{6}, & 4 \leq x < 6 \\
 \frac{1}{3}, & x=6 \\
 \frac{10-x}{12}, & 6 < x \leq 10 \\
 0, & \text{otherwise}
 \end{cases}$$

3. To verify that $f_X$ qualifies as a valid probability density function, we must confirm that its total area equals $1$ by evaluating the integral over its support:

$$\int_{4}^{10} f_X(x)\,dx = \int_{4}^{6} \frac{x-4}{6}\,dx + \int_{6}^{10} \frac{10-x}{12}\,dx
$$

Evaluating the first integral (left slope):

$$\frac{1}{6} \left[ \frac{(x-4)^2}{2} \right]_{4}^{6} = \frac{1}{6} \left( \frac{2^2}{2} - 0 \right) = \frac{2}{6} = \frac{1}{3}$$

Evaluating the second integral (right slope):

$$\frac{1}{12} \left[ -\frac{(10-x)^2}{2} \right]_{6}^{10} = \frac{1}{12} \left( 0 - \left( -\frac{4^2}{2} \right) \right) = \frac{8}{12} = \frac{2}{3}$$

Combining these results, we find that $\frac{1}{3} + \frac{2}{3} = 1$, confirming that $f_X$ is a legitimate density function.

4. The Expected Value for a Triangular distribution is the simple average of its three parameters: the lower bound ($a$), the upper bound ($b$), and the mode ($c$):

$$E[X] = \frac{a+b+c}{3} = \frac{4+10+6}{3} = \frac{20}{3} \approx 6.67$$

The Variance is determined by the spread and internal positioning of these parameters using the formula:

$$\text{Var}(X) = \frac{a^2+b^2+c^2 - ab - ac - bc}{18}$$

Substituting our specific values ($a=4, b=10, c=6$):

$$\text{Var}(X) = \frac{16+100+36 - 40 - 24 - 60}{18} = \frac{28}{18}$$

This simplifies to $\frac{14}{9} \approx 1.56$.

6. In this context:

* $a=4$ is the shortest possible installation time,
* $b=10$ is the longest possible installation time,
* $c=6$ is the most likely installation time.

7.Given the lack of historical data, a Triangular distribution is the most suitable model. It captures the essential 'best-case, worst-case, and most-likely' judgments of subject matter experts, providing a clear, bounded framework for our analysis.

8. To calculate the probability that the installation takes 5 hours or less, $P(X \leq 5)$, we integrate the density function over the interval $[4, 5]$. Since this interval falls within the first branch of the distribution ($4 \leq x < 6$), the calculation is as follows:

$$P(X \leq 5) = \int_{4}^{5} \frac{x-4}{6} \, dx$$Extracting the constant and integrating:$$P(X \leq 5) = \frac{1}{6} \left[ \frac{(x-4)^2}{2} \right]_{4}^{5} = \frac{1}{6} \left( \frac{1^2}{2} - 0 \right) = \frac{1}{12}$$

Thus, $P(X \leq 5) \approx 0.0833$, indicating an $8.33\%$ chance the task is completed within 5 hours.

9. To determine the probability that the installation takes between 5 and 8 hours, $P(5 \leq X \leq 8)$, we must integrate over both branches of the density function, splitting the calculation at the mode ($x=6$):

$$P(5 \leq X \leq 8) = \int_{5}^{6} \frac{x-4}{6}\,dx + \int_{6}^{8} \frac{10-x}{12}\,dx$$First Segment (Increasing Slope):

$$\frac{1}{6} \left[ \frac{(x-4)^2}{2} \right]_{5}^{6} = \frac{1}{6} \left( \frac{4}{2} - \frac{1}{2} \right) = \frac{3}{12} = \frac{1}{4}$$Second Segment (Decreasing Slope):

$$\frac{1}{12} \left[ -\frac{(10-x)^2}{2} \right]_{6}^{8} = \frac{1}{12} \left( -\frac{4}{2} - (-\frac{16}{2}) \right) = \frac{6}{12} = \frac{1}{2}$$

Summing these results gives $P(5 \leq X \leq 8) = \frac{1}{4} + \frac{1}{2} = \mathbf{0.75}$. There is a $75\%$ chance the installation will fall within this specific time window.

10. By defining the probability of completion within a target window, this model enhances the accuracy of project timelines. It also supports sensitivity testing by modeling how variance in individual task durations influences the critical path, thereby optimizing resource deployment and risk mitigation strategies..


```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
a = 4
b = 10
c_mode = 6

# Generate x values
x = np.linspace(0, b, 500)

# Triangular PDF (computed directly, no scipy needed)
def triangular_pdf(x, a, b, c):
    pdf = np.where(
        x < c,
        (2 * (x - a)) / ((b - a) * (c - a)),
        np.where(
            x <= b,
            (2 * (b - x)) / ((b - a) * (b - c)),
            0.0
        )
    )
    pdf = np.where(x < a, 0.0, pdf)
    return pdf

y = triangular_pdf(x, a, b, c_mode)

# Plot
fig, ax = plt.subplots(figsize=(9, 5))

ax.plot(x, y, color='royalblue', linewidth=2, label='Triangular Distribution')
ax.fill_between(x, y, alpha=0.25, color='royalblue')

ax.set_title(f'Triangular Distribution (a={a}, b={b}, c={c_mode})', fontsize=14)
ax.set_xlabel('Time (hours)')
ax.set_ylabel('Probability Density')
ax.set_xlim(0, b)
ax.set_ylim(bottom=0)
ax.legend()
ax.grid(True, linestyle='--', alpha=0.4)

plt.tight_layout()
plt.show()
```

1. Given the project estimates for the minimum ($a=8$), maximum ($b=20$), and most likely ($c=11$) durations, we can derive the shape parameters for the PERT distribution.The first shape parameter ($\alpha$) is calculated based on the distance from the minimum to the mode:

$$\alpha = 1 + 4\left( \frac{11-8}{20-8} \right) = 1 + 4\left( \frac{3}{12} \right) = 2$$

The second shape parameter ($\beta$) is determined by the distance from the mode to the maximum:

$$\beta = 1 + 4\left( \frac{20-11}{20-8} \right) = 1 + 4\left( \frac{9}{12} \right) = 4$$

These parameters, $\alpha=2$ and $\beta=4$, define the specific curvature and skewness of the PERT distribution for this task.

2. Since $X$ represents the task duration in days, its support is

$$

3. The probability density function (PDF) for a PERT-distributed random variable $X$ on the interval $[a, b]$ is defined as:

$$f_X(x) = \frac{(x-a)^{\alpha-1}(b-x)^{\beta-1}}{B(\alpha, \beta)(b-a)^{\alpha+\beta-1}}, \quad a \leq x \leq b$$

Substituting the specific parameters $a=8, b=20, \alpha=2,$ and $\beta=4$, the expression becomes:

$$f_X(x) = \frac{(x-8)^{1}(20-x)^{3}}{B(2,4) \cdot 12^5}$$

To finalize the density, we evaluate the Beta function using the relationship with Gamma functions:

$$B(2,4) = \frac{\Gamma(2)\Gamma(4)}{\Gamma(2+4)} = \frac{1! \cdot 3!}{5!} = \frac{1 \cdot 6}{120} = \frac{1}{20}$$

Incorporating this constant back into the PDF yields the simplified density:
$$f_X(x) = \begin{cases}
\frac{20(x-8)(20-x)^3}{12^5}, & 8 \leq x \leq 20 \\ 0, &
\text{otherwise}
\end{cases}
$$

4. To elevate the formal mathematical rigor of your document, here are three ways to rewrite this definition of the probability space:1. The Formal Mathematical Version"The task duration is modeled on a probability space $(\Omega, \mathfrak{F}, P)$, where:

6. To derive the expected value $E[X]$, we evaluate the first moment of the PERT density function. The integration is simplified by mapping the support $[a, b]$ to the standard unit interval $[0, 1]$ using the transformation $t = \frac{x-a}{b-a}$.

$$E[X] = \int_{0}^{1} \big( a + (b-a)t \big) \frac{t^{\alpha-1}(1-t)^{\beta-1}}{B(\alpha,\beta)} \, dt$$

Distributing the terms allows us to express the integral in terms of Beta functions:

$$E[X] = a \frac{B(\alpha, \beta)}{B(\alpha, \beta)} + (b-a) \frac{B(\alpha+1, \beta)}{B(\alpha, \beta)} = a + (b-a) \frac{\alpha}{\alpha+\beta}$$

Given the PERT definitions for $\alpha$ and $\beta$, their sum is constant at $\alpha + \beta = 6$. Substituting $\alpha = 1 + 4\frac{c-a}{b-a}$ into the expectation formula yields the classic PERT weighted average:

$$E[X] = \frac{a + 4c + b}{6}$$

Applying our parameters ($a=8, b=20, c=11$):

$$E[X] = \frac{8 + 44 + 20}{6} = \frac{72}{6} = 12 \text{ days.}$$

7. The variance of a PERT distribution is commonly estimated using the relationship between the mean ($E[X]$) and the boundary parameters ($a$ and $b$). The specific approximation formula is:

$$\text{Var}(X) = \frac{(E[X] - a)(b - E[X])}{7}$$

Substituting the previously calculated mean of $12$ and the bounds $a=8, b=20$:

$$\text{Var}(X) = \frac{(12-8)(20-12)}{7} = \frac{4 \cdot 8}{7} = \frac{32}{7}$$

This yields a variance of approximately $4.57$, providing a measure of the uncertainty inherent in the task duration.

8. The standard deviation is

$$
\sigma=\sqrt{\mathrm{Var}(X)}
=\sqrt{\frac{32}{7}}
\approx 2.14.

Using the rule of thumb,

$$
\sigma \approx \frac{b-a}{6}

which is close to the value above.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import triang, beta

# Parameters
a = 4
b = 10
c_mode = 6

# --- Triangular Distribution ---
c_scipy = (c_mode - a) / (b - a)
loc = a
scale = b - a

# --- PERT Distribution (Beta with PERT parameters) ---
alpha_pert = 1 + 4 * (c_mode - a) / (b - a)
beta_pert  = 1 + 4 * (b - c_mode) / (b - a)

# Generate data points
x = np.linspace(0, b, 500)
y_triang = triang.pdf(x, c=c_scipy, loc=loc, scale=scale)
y_pert   = beta.pdf(x, alpha_pert, beta_pert, loc=a, scale=b - a)

# Plot
fig, ax = plt.subplots(figsize=(9, 5))

ax.plot(x, y_triang, color='royalblue', linewidth=2, linestyle='--', label='Triangular Distribution')
ax.plot(x, y_pert,   color='firebrick', linewidth=2,                  label='PERT Distribution')

ax.set_title(f'Comparison: Triangular vs PERT (a={a}, b={b}, c={c_mode})', fontsize=14)
ax.set_xlabel('Time (hours)')
ax.set_ylabel('Probability Density')
ax.set_xlim(0, b)
ax.set_ylim(bottom=0)
ax.legend()
ax.grid(True, linestyle='--', alpha=0.4)

plt.tight_layout()
plt.show()
```
